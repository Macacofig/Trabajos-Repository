La llegada de la Inteligencia Artificial generativa al desarrollo de software ha introducido una capa fascinante —y peligrosa— de nuevos sesgos y falacias. Ahora no solo debemos cuidarnos de nuestra propia mente, sino también de cómo interpretamos lo que nos dice la "máquina".

Aquí tienes ejemplos aplicados específicamente al uso de herramientas como ChatGPT, Claude, Gemini o Copilot:

---

## 1. Nuevos Sesgos en la Era de la IA

### Sesgo de Automatización (Automation Bias)

Es la tendencia a confiar excesivamente en las sugerencias de sistemas automatizados, incluso cuando contradicen nuestro propio juicio o datos reales.

* **En el software:** Un desarrollador le pide a la IA que genere una función criptográfica o un middleware de autenticación. La IA lo genera con una sintaxis impecable, y el desarrollador lo pega sin auditarlo, asumiendo que "si la IA lo hizo, debe ser seguro".
* **Riesgo:** Introducción de vulnerabilidades de seguridad graves o lógica de negocio errónea que parece correcta a simple vista.

### Sesgo de Anclaje (Anchoring Bias)

Ocurre cuando la primera información que recibimos domina nuestro pensamiento posterior.

* **En el software:** Pides a la IA una estructura para una nueva base de datos. La IA te da una opción basada en SQL. A partir de ahí, todo tu diseño gira en torno a SQL, incluso si tras analizarlo más a fondo, una base de datos de grafos o documental hubiera sido más eficiente. Te "anclaste" a la primera respuesta de la IA.

### Sesgo de Disponibilidad

Tendencia a valorar más la información que la IA nos da "a la mano" simplemente porque fue fácil de obtener, ignorando soluciones que requerirían una investigación más profunda en documentación oficial.

* **En el software:** Usar una librería sugerida por la IA que está obsoleta o tiene poco mantenimiento, solo porque fue el ejemplo de código que la IA generó rápidamente.

---

## 2. Falacias Lógicas aplicadas a la IA

### Falacia Ad Populum (Basada en el entrenamiento)

Argumentar que algo es la "mejor práctica" simplemente porque la IA (que ha leído millones de repositorios) lo sugiere.

* **Ejemplo:** "La IA me sugirió usar este patrón de diseño complejo para este pequeño script, así que debe ser el estándar de la industria".
* **Realidad:** La IA predice el próximo token basado en patrones estadísticos de la web; si la mayoría de la gente escribe código mediocre en GitHub, la IA podría sugerir patrones mediocres por pura probabilidad.

### Falacia del Francotirador (Texas Sharpshooter)

Ocurre cuando se seleccionan solo los datos que confirman una conclusión y se ignora el resto.

* **Ejemplo:** Un desarrollador prueba un código generado por IA, falla 4 veces, pero a la quinta funciona. Entonces dice: "Esta IA es infalible para arquitectura en la nube", ignorando los cuatro fallos previos.
* **Realidad:** Se está forzando una narrativa de éxito ignorando la tasa de error real del modelo.

### Falacia de la Falsa Equivalencia

Creer que porque una IA es excelente explicando conceptos teóricos de algoritmos, tiene la misma capacidad para entender el contexto de negocio específico de tu empresa.

* **Ejemplo:** "Si la IA pudo resolver este problema de LeetCode de nivel difícil, seguramente su sugerencia sobre cómo manejar nuestros pagos internacionales es correcta".
* **Realidad:** Resolver un puzzle lógico es muy distinto a entender las sutilezas legales y de negocio de un dominio específico.

---

## Cómo mitigar esto: El Ciclo de Retroalimentación Humana

Para no caer en estas trampas, es vital aplicar un enfoque de **"Humano en el bucle" (Human-in-the-loop)**.

### Buenas prácticas contra el "Sesgo de IA":

1. **Principio de Desconfianza Inicial:** Trata cada salida de la IA como un "borrador de un junior entusiasta". Siempre debe pasar por un Code Review humano.
2. **Prompting Adversarial:** En lugar de pedirle que "escriba el código", pídele que "encuentre vulnerabilidades en este código que ella misma escribió". Esto ayuda a romper el sesgo de confirmación.
3. **Verificación de Hechos (Fact-checking):** Si la IA cita una librería o una función específica, verifica en la documentación oficial que realmente existe y no es una "alucinación".



Esta es una excelente iniciativa. El objetivo no es prohibir la IA, sino establecer una **"Aduana Mental"** que filtre lo útil de lo peligroso.

Aquí tienes un Checklist diseñado para ser integrado en tus *Pull Requests* (PR) o como guía rápida para los desarrolladores cuando trabajan con asistentes como Gemini, ChatGPT o Copilot.

---

## ⚠️ Checklist de Validación: Código Generado por IA

### 1. Integridad y Seguridad (Lo más crítico)

* [ ] **¿Entiendo cada línea?** No apruebes código que no podrías explicar línea por línea en una reunión técnica.
* [ ] **¿Hay secretos expuestos?** Revisa que la IA no haya incluido claves API falsas, tokens o URLs de entornos de prueba que puedan filtrarse.
* [ ] **Vulnerabilidades comunes:** ¿El código introdujo riesgos de SQL Injection, Cross-Site Scripting (XSS) o falta de validación de entradas?
* [ ] **Criptografía:** Si el código maneja seguridad, ¿usa librerías modernas o se inventó un algoritmo propio/obsoleto?

### 2. Contexto y Arquitectura

* [ ] **Cumplimiento de Estándares:** ¿Sigue el código las convenciones de nombres y la arquitectura de nuestro proyecto (ej. Clean Architecture, Hexagonal) o ha introducido un estilo "genérico"?
* [ ] **Dependencias Existentes:** ¿Está usando las librerías que ya tenemos en el proyecto o sugirió instalar una nueva innecesariamente?
* [ ] **Escalabilidad:** ¿El algoritmo funciona bien con 10 registros pero fallará con 10.000? (La IA suele priorizar la brevedad sobre el rendimiento).

### 3. Veracidad (Detección de Alucinaciones)

* [ ] **Existencia de APIs/Librerías:** ¿Las funciones y métodos invocados realmente existen en la versión de la librería que usamos? (Cuidado con versiones inventadas).
* [ ] **Lógica de Negocio:** ¿El código respeta nuestras reglas de negocio específicas o está aplicando una lógica "estándar" que no aplica aquí?

### 4. Mantenibilidad y Testing

* [ ] **Legibilidad:** ¿El código es "humano"? A veces la IA genera código demasiado complejo o comprimido (*clever code*) que es difícil de mantener.
* [ ] **Unit Tests:** ¿Le pediste a la IA que genere también los tests? Si es así, ¿los tests realmente prueban la lógica o son tests redundantes que siempre pasan?
* [ ] **Comentarios:** ¿Los comentarios explican el "por qué" o simplemente repiten lo que hace el código?

---

## 🛠️ Protocolo de Acción Sugerido

Para que este checklist sea efectivo, recomiendo implementar la regla de los **"Dos Pares de Ojos"**:

1. **Ojo 1 (El Desarrollador):** Pasa el checklist antes de subir el código.
2. **Ojo 2 (El Revisor):** En el proceso de Code Review, si se detecta que el código fue generado por IA, el revisor debe ser **doblemente escéptico**.

> **Pro-Tip de Prompting:** Una forma de automatizar parte de esto es pasarle tu código final a la IA con el siguiente prompt: *"Actúa como un Senior Security Engineer y un experto en Performance. Encuentra 3 fallos potenciales y 2 áreas de mejora en este código que acabas de generar"*. Esto ayuda a romper el **Sesgo de Automatización**.
