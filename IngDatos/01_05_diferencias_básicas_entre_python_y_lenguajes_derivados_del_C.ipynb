## Type Hints en Python

### 1. Definición

Introducidos en Python 3.5 (vía PEP 484), los **Type Hints** son una sintaxis que te permite indicar explícitamente el tipo de dato que se espera en una variable, un parámetro de función o el valor de retorno.

Es fundamental entender que **Python sigue siendo un lenguaje de tipado dinámico**. Esto significa que el intérprete de Python no te obligará a cumplir con estos tipos en tiempo de ejecución. Son "pistas" que sirven principalmente para:

* **Herramientas de análisis estático** (como Mypy).
* **Autocompletado en IDEs** (como VS Code o PyCharm).
* **Documentación del código.**

### 2. Ejemplos

#### Variables simples

```python
nombre: str = "Alice"
edad: int = 30
precio: float = 99.99
es_activo: bool = True

print(f"Nombre: {nombre}, Tipo: {type(nombre)}")
print(f"Edad: {edad}, Tipo: {type(edad)}")
print(f"Precio: {precio}, Tipo: {type(precio)}")
print(f"¿Está activo?: {es_activo}, Tipo: {type(es_activo)}")
```

#### Argumentos de funciones y valores de retorno

```python
def saludar(nombre: str) -> str:
    return f"Hola, {nombre}!"

def sumar(a: int, b: int) -> int:
    return a + b

print(saludar("Bob"))
print(sumar(5, 7))
```

#### Colecciones (Listas, Tuplas, Diccionarios, Sets)

```python
# from typing import List, Dict, Tuple, Set
# En versiones modernas (Python 3.9+), puedes usar las colecciones estándar directamente:

numeros: list[int] = [1, 2, 3, 4]
personas: dict[str, int] = {"Alice": 30, "Bob": 25}
coordenadas: tuple[float, float] = (10.5, 20.3)
frutas: set[str] = {"manzana", "pera", "uva"}

print(f"Números: {numeros}, Tipo: {type(numeros)}")
print(f"Personas: {personas}, Tipo: {type(personas)}")
print(f"Coordenadas: {coordenadas}, Tipo: {type(coordenadas)}")
print(f"Frutas: {frutas}, Tipo: {type(frutas)}")
```

#### Clases y Objetos

```python
class Punto:
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y

def mover_punto(p: Punto, dx: float, dy: float) -> Punto:
    p.x += dx
    p.y += dy
    return p

mi_punto: Punto = Punto(1.0, 2.0)
print(f"Punto original: ({mi_punto.x}, {mi_punto.y})")

punto_movido = mover_punto(mi_punto, 0.5, -1.0)
print(f"Punto movido: ({punto_movido.x}, {punto_movido.y})")
```

#### Tipos Opcionales o Múltiples (Python 3.10+)

```python
from typing import Optional

# Optional[str] = str | None
def buscar_usuario(id: int) -> str | None:
    if id == 1:
        return "Admin"
    return None
```

#### Tipos de librerías

```python
import pandas as pd

def procesar_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    """
    Esta función toma un DataFrame, añade una columna nueva y devuelve el DataFrame modificado.
    """
    # Se crea una nueva columna con el doble de los valores de una columna existente (si existe)
    if 'valor' in df.columns:
        df['valor_doble'] = df['valor'] * 2
    else:
        print("Advertencia: La columna 'valor' no se encontró en el DataFrame.")
        df['valor_doble'] = 0 # O alguna lógica predeterminada
    return df

# Crear un DataFrame de ejemplo
datos = {'id': [1, 2, 3], 'valor': [10, 20, 30]}
df_original = pd.DataFrame(datos)
print("DataFrame original:")
print(df_original)

# Llamar a la función con el type hint de pd.DataFrame
df_procesado = procesar_dataframe(df_original)
print("\nDataFrame procesado:")
print(df_procesado)

# Ejemplo de cómo un type checker podría identificar un posible error si pasamos el tipo incorrecto
# Aunque Python no lo forzará en tiempo de ejecución, un linter o mypy sí lo harían.
# def otra_funcion(data: pd.DataFrame) -> None:
#     print(data.head())
#
# otra_funcion("esto no es un dataframe") # Un type checker advertiría aquí
```

### 3. Casos de Uso

*   **Mejora de la legibilidad y auto-documentación**: El código con Type Hints es más fácil de entender para otros desarrolladores (y para ti mismo en el futuro), ya que las intenciones de tipos son explícitas.
*   **Detección de errores temprana**: Herramientas como `mypy` pueden analizar tu código e identificar inconsistencias de tipo antes de que se ejecute, lo que reduce el tiempo de depuración.
*   **Mejor soporte IDE**: Entornos de desarrollo integrados (IDEs) como VS Code o PyCharm utilizan Type Hints para proporcionar autocompletado más preciso, refactorización inteligente y detección de errores en tiempo real.
*   **Facilita el mantenimiento**: A medida que los proyectos crecen y evolucionan, los Type Hints ayudan a asegurar que los cambios de código no introduzcan errores de tipo inesperados en otras partes del sistema.
*   **Diseño de APIs claras**: Al definir funciones y clases, los Type Hints son excelentes para comunicar cómo se deben usar correctamente tus componentes.

### 4. Buenas Prácticas

1.  **Sé consistente**: Decide si usarás Type Hints en tu proyecto y aplícalos consistentemente. La inconsistencia puede ser más confusa que no usarlos en absoluto.
2.  **Usa `mypy` u otros linters**: Integra un analizador estático de tipos en tu flujo de trabajo de desarrollo (por ejemplo, en tus hooks de pre-commit o CI/CD) para aprovechar al máximo los Type Hints.
3.  **Empieza gradualmente**: No es necesario tipar todo un proyecto existente de una vez. Puedes empezar con las partes más críticas o las funciones/módulos nuevos.
4.  **No tipes todo:** No es necesario tipar variables locales simples donde el tipo es obvio (ej. `x = 10`). Enfócate en las **firmas de las funciones** y **clases**.
5.  **Utiliza `typing.Any` cuando sea necesario**: Si tienes un tipo que realmente puede ser cualquier cosa y no puedes ser más específico, puedes usar `from typing import Any`. Esto le dice al analizador de tipos que no imponga restricciones en ese lugar.
6.  **Comprende la diferencia con el tipado estático**: Recuerda que Python sigue siendo dinámico. Los Type Hints son solo sugerencias; el intérprete no las hace cumplir en tiempo de ejecución. Esto significa que puedes "engañar" a los Type Hints, aunque hacerlo anula su propósito.

### Resumen

| Característica | Sin Type Hints | Con Type Hints |
| --- | --- | --- |
| **Claridad** | Media (depende de nombres) | Alta (explícita) |
| **Errores** | Se detectan al ejecutar | Se detectan al escribir |
| **IDE** | Sugerencias genéricas | Sugerencias precisas |
