# Iterables

En Python, un iterable es cualquier objeto del que puedes obtener elementos uno por uno. Piensa en ello como una colección de elementos que se pueden recorrer o "iterar". Si puedes usar un bucle for para recorrer los elementos de un objeto, entonces ese objeto es un iterable.

Algunos ejemplos comunes de iterables en Python son:

- **Listas** (como ['apple', 'banana', 'cherry'])
- **Tuplas** (como ('red', 'green', 'blue'))
- **Cadenas de texto** (strings) (como 'hello')
- **Rangos** (como range(0, 5))
- **Diccionarios** (iteran sobre sus claves o valores)
- **Conjuntos** (sets)

## Secuencias

Una **secuencia** en Python es un tipo de iterable que tiene un **orden definido** y cada elemento dentro de ella puede ser accedido por un índice numérico (empezando desde 0). Esto significa que sus elementos están ordenados y puedes saber la posición exacta de cada uno.

Las características clave de las secuencias son:

- **Ordenado**: Los elementos mantienen un orden específico en el que fueron agregados.
- **Indexable**: Puedes acceder a los elementos usando un índice numérico (por ejemplo, `mi_lista[0]`).
- **Slicing**: Puedes extraer sub-secuencias (partes de la secuencia) usando la notación de `[inicio:fin:paso]`.

Ejemplos comunes de secuencias en Python incluyen:

- **Listas** (list): Son secuencias mutables (se pueden cambiar).
- **Tuplas** (tuple): Son secuencias inmutables (no se pueden cambiar una vez creadas).
- **Cadenas de texto** (str): Son secuencias inmutables de caracteres.
- **Rangos** (range): Son secuencias inmutables de números.

Todas las secuencias son iterables porque puedes recorrer sus elementos uno por uno. Sin embargo, no todos los iterables son secuencias. Por ejemplo, los conjuntos (set) y los diccionarios (dict) son iterables, pero no son secuencias porque sus elementos no tienen un orden definido ni se acceden por índice numérico.

## Manipulación de listas

Referencia: https://docs.python.org/3.12/tutorial/datastructures.html#more-on-lists

````python
marcas_automoviles = ['toyota','ford','fiat']

#Agregar elementos
print(marcas_automoviles)
marcas_automoviles.append('ferrari')
print(marcas_automoviles)
marcas_automoviles.insert(0,'audi')
print(marcas_automoviles)

#Modificar elementos
marcas_automoviles[4] = 'mazda'
print(marcas_automoviles)

#Eliminar elementos
del marcas_automoviles[4]
print(marcas_automoviles)
marcas_automoviles.remove('ford')
print(marcas_automoviles)
print(marcas_automoviles.pop(0))
print(marcas_automoviles)

#Buscar un elemento
print(marcas_automoviles.index('fiat'))

#Otras operaciones
copia_marcas = marcas_automoviles.copy()
marcas_automoviles.clear()
print(marcas_automoviles)
print(copia_marcas)
````

## __[List Comprehensions](https://realpython.com/list-comprehension-python/)__

Objetivo: generar una lista a partir de la transformación y opcionalmente la selección de los elementos de una secuencia.

````python
marcas_automoviles
````

````python
marcas_automoviles = ['DYB', 'toyota', 'ford', 'fiat', 'ferrari']
````

````python
marcas_mays = []
for marca in marcas_automoviles:
    if len(marca) >= 4:
       marcas_mays.append(len(marca))

print(marcas_mays)
````

````python
# Conversión del código en celda anterior a un list comprehension
#[ expresion for-loop expresion-filtro ]
marcas_mays = [ len(marca)
                for marca in marcas_automoviles
                if len(marca) >= 4 ]
marcas_mays
````

````python
[ marca.upper() for marca in marcas_automoviles if len(marca) > 4]
````

````python
[marca.upper() for marca in marcas_automoviles if len(marca) > 4]
````

````python
marcas_mays = []
for marca in marcas_automoviles:
  if len(marca) > 4:
    marcas_mays.append(marca.upper())

print(marcas_mays)
````

````python
# Cómo organizart el código para que cada sección de un list comprehension esté en su propia línea.
[marca.upper()
for marca in marcas_automoviles
if len(marca) > 4]
````
