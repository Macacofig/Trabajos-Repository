# Prompts - Role-Goal-Context

La estrategia **<font color="#8FC07C">Role</font>-<font color="#7C8FC0">Goal</font>-<font color="#C07C8F">Context</font>** (RGC) estructura solicitudes (prompts) definiendo claramente tres elementos clave:

**Role (Rol)** se trata de establecer la perspectiva o persona que quieres que Gemini adopte. Esto le dice a Gemini quién debe "ser" o qué perspectiva debe tomar frente al problema. Por ejemplo, podrías pedirle a Gemini que responda como un científico de datos, un editor de escritura creativa, un programador principiante o un experto en un dominio. Esto ayuda a Gemini a adaptar su lenguaje, profundidad de explicación y enfoque para coincidir con esa perspectiva.

**Goal (Objetivo)** es el resultado específico o tarea que quieres lograr. Este es tu objetivo principal—lo que quieres que Gemini produzca o resuelva. Debe ser claro y mensurable cuando sea posible. Los ejemplos incluyen "escribir un correo electrónico persuasivo," "explicar este concepto de forma simple," "depurar este código," o "analizar este conjunto de datos para buscar tendencias."

**Context (Contexto)** proporciona la información de antecedentes y las restricciones que dan forma a la tarea. Esto incluye detalles relevantes sobre la situación, cualquier restricción dentro de la cual Gemini debe trabajar, la audiencia objetivo, el formato deseado, tono, *ejemplos variados*, longitud u otras especificaciones. El contexto ayuda a Gemini a entender por qué importa la tarea y cómo abordarla mejor.

**Un ejemplo práctico:** En lugar de preguntar "Escribe sobre el cambio climático," usarías RGC así:

"<font color="#C07C8F">Como <font color="#8FC07C">científico ambiental que se comunica con legisladores</font>, ayúdame a <font color="#7C8FC0">escribir un documento informativo</font> de 2 páginas sobre los impactos económicos del cambio climático. Enfócate en información procesable e incluye puntos de datos específicos que influirían en las decisiones políticas.</font>"

Aquí el rol es un científico ambiental, el objetivo es escribir un documento de política, y el contexto incluye la audiencia, longitud, tono y áreas clave de enfoque.

Esta estrategia funciona bien porque reduce la ambigüedad y ayuda a Gemini a entender exactamente lo que necesitas.

## Named Tuples en Python

Las `namedtuple`s son una forma de crear subtipos de tuplas con campos nombrados. Son una característica del módulo `collections` en Python. Ofrecen la inmutabilidad y eficiencia de las tuplas regulares, pero con una sintaxis más legible y auto-documentada para acceder a los elementos.

### Definición

Una `namedtuple` es esencialmente una tupla inmutable donde puedes acceder a sus elementos tanto por índice (como en una tupla normal) como por un nombre de campo (como en un diccionario o un objeto con atributos).

`namedtuple` es una función factoría para crear clases de tuplas personalizadas.

### Ejemplos

Vamos a ver cómo se define y se usa una `namedtuple`.

```python
from collections import namedtuple

# 1. Definición de una namedtuple
# NamedTuple('NombreDeLaClase', ['campo1', 'campo2', ...])

# Creamos una namedtuple para representar un punto en 2D
Point = namedtuple('Point', ['x', 'y'])

# Creamos instancias de Point
p1 = Point(10, 20)
p2 = Point(x=30, y=40)

print(f"Punto p1: {p1}")
print(f"Punto p2: {p2}")

# Acceso a los elementos
print("\nAcceso por índice:")
print(f"p1.x (índice 0): {p1[0]}")
print(f"p1.y (índice 1): {p1[1]}")

print("\nAcceso por nombre de campo:")
print(f"p2.x: {p2.x}")
print(f"p2.y: {p2.y}")

# Las namedtuples son inmutables (como las tuplas)
try:
    p1.x = 50
except AttributeError as e:
    print(f"\nError al intentar modificar un campo: {e}")
```

Podemos definir `namedtuple`s con más campos y también con valores por defecto.

```python
from collections import namedtuple

# Namedtuple para representar un libro
Book = namedtuple('Book', ['title', 'author', 'year', 'isbn', 'pages'])

# Creamos instancias de Book
book1 = Book('Cien años de soledad', 'Gabriel García Márquez', 1967, '978-0307474278', 417)
book2 = Book('1984', 'George Orwell', 1949, '978-0451524935', 328)

print(f"\nLibro 1: {book1}")
print(f"Título del libro 1: {book1.title}")
print(f"Autor del libro 2: {book2.author}")

# Namedtuple con valores por defecto (requiere sintaxis _make y _asdict)
# Una forma más sencilla para valores por defecto es usar el parámetro defaults

# Python 3.7+ permite defaults directamente
Person = namedtuple('Person', ['name', 'age', 'city'], defaults=['Unknown'])

p = Person('Alice', 30)
print(f"\nPersona p: {p}")

p_with_city = Person('Bob', 25, 'New York')
print(f"Persona p_with_city: {p_with_city}")

# Métodos útiles
# _make(): crea una nueva instancia a partir de una secuencia iterable
data = ['Charlie', 40, 'London']
p_from_make = Person._make(data)
print(f"Persona creada con _make: {p_from_make}")

# _asdict(): devuelve un diccionario ordenado con los campos y sus valores
person_dict = p_with_city._asdict()
print(f"Persona como diccionario: {person_dict}")
print(f"Tipo de person_dict: {type(person_dict)}")

# _replace(): crea una nueva namedtuple con algunos campos reemplazados
p_updated = p_with_city._replace(age=26)
print(f"Persona actualizada con _replace: {p_updated}")
print(f"Original p_with_city: {p_with_city}") # Se mantiene inmutable
```

### Casos de Uso y Buenas Prácticas

Las `namedtuple`s son excelentes para mejorar la legibilidad del código y la auto-documentación de estructuras de datos inmutables.

#### Casos de Uso:

1.  **Representar registros o filas de datos**: Cuando tienes una colección de datos con un esquema fijo, como filas de una base de datos o CSV, o resultados de una API. Permiten acceder a los campos por nombre, lo que es mucho más claro que por índice (e.g., `registro.nombre` vs `registro[0]`).
2.  **Mejorar la legibilidad de tuplas**: Cuando pasas tuplas como argumentos o las devuelves de funciones, una `namedtuple` hace que el código sea más legible y menos propenso a errores. En lugar de `def process_coords(coords): x = coords[0]`, puedes tener `def process_point(point): x = point.x`.
3.  **Objetos ligeros e inmutables**: Son más ligeras en memoria que los objetos de clases personalizadas (especialmente si se usa `__slots__`) y garantizan la inmutabilidad. Son ideales cuando necesitas una estructura de datos simple que no requiere métodos complejos ni herencia.
4.  **Alternativa a diccionarios para datos fijos**: Si la estructura de los datos es conocida y fija, una `namedtuple` es más eficiente y segura que un diccionario, ya que no puedes acceder a una clave inexistente y los nombres de los campos están definidos en la clase.
5.  **Simulacros (Mocks) en pruebas**: Pueden ser útiles para crear objetos de prueba simples y predecibles.

#### Buenas Prácticas:

1.  **Nombres claros para campos y clase**: Utiliza nombres descriptivos y sigue las convenciones de nomenclatura de Python (PascalCase para la clase de la `namedtuple`, snake_case para los campos).
2.  **Cuando la inmutabilidad es clave**: Úsalas cuando los datos representados no deban cambiar después de la creación. Si necesitas un objeto mutable, una clase estándar es más apropiada.
3.  **Evita el uso excesivo de `_make`, `_asdict`, `_replace`**: Aunque útiles, estos métodos son principalmente para la manipulación programática. Para la interacción diaria, el acceso por atributo (`.`) es la forma preferida.
4.  **No las uses para herencia compleja**: Las `namedtuple`s no están diseñadas para ser extendidas a través de una herencia compleja. Si necesitas una jerarquía de clases o métodos complejos, define una clase regular.
5.  **Considera `dataclasses` para más funcionalidad**: Si necesitas características como valores por defecto más avanzados, validación de tipos, o la capacidad de añadir métodos fácilmente, las `dataclasses` (introducidas en Python 3.7) suelen ser una opción más moderna y potente. Las `namedtuple`s siguen siendo excelentes para casos de uso muy simples y ligeros.

En resumen, las `namedtuple`s son una herramienta poderosa para escribir código más legible y robusto al trabajar con colecciones de datos estructurados e inmutables.

## Pandas vs Polars: Principales Diferencias y Migración

### Principales Diferencias

**Pandas**:
*   **Ejecución:** Single-threaded (monohilo) por defecto, aunque puede usar múltiples núcleos en algunas operaciones a través de librerías externas o optimizaciones. Se enfoca en el procesamiento en memoria (in-memory processing).
*   **Modelo de Datos:** Utiliza NumPy como su base, con DataFrames mutables. Esto significa que las operaciones pueden modificar el DataFrame original.
*   **API:** Muy madura y ampliamente adoptada, con una gran cantidad de funcionalidades y una vasta comunidad.
*   **Rendimiento:** Puede ser lento en grandes conjuntos de datos debido a su naturaleza monohilo y el uso de copias de datos en ciertas operaciones.
*   **Sintaxis:** Generalmente más imperativa, con operaciones que a menudo se aplican en secuencia.

**Polars**:
*   **Ejecución:** Altamente paralelizada y optimizada para el rendimiento, aprovechando todos los núcleos de la CPU. Utiliza Apache Arrow como su base, lo que permite un procesamiento de datos eficiente y cero copias.
*   **Modelo de Datos:** DataFrames inmutables (por defecto). Las operaciones devuelven nuevos DataFrames, lo que puede simplificar el razonamiento sobre el flujo de datos. Está diseñado para el procesamiento "out-of-core" (manejo de datos más grandes que la RAM disponible).
*   **API:** Más moderna y diseñada con el rendimiento en mente. Ofrece una sintaxis expresiva y optimizada para operaciones en cadena (chaining).
*   **Rendimiento:** Generalmente mucho más rápido que Pandas, especialmente con grandes volúmenes de datos y operaciones complejas, gracias a su paralelismo y optimización de memoria.
*   **Sintaxis:** A menudo más declarativa y funcional, promoviendo el encadenamiento de métodos.

### Cambios para Migrar de Pandas a Polars

La migración de Pandas a Polars implica adaptar algunas construcciones y paradigmas. Aquí se destacan los cambios clave:

1.  **Importación:**
    *   **Pandas:** `import pandas as pd`
    *   **Polars:** `import polars as pl`

2.  **Creación de DataFrames:**
    *   **Pandas:** `df_pd = pd.DataFrame({'col1': [1, 2], 'col2': [3, 4]})`
    *   **Polars:** `df_pl = pl.DataFrame({'col1': [1, 2], 'col2': [3, 4]})`

3.  **Lectura de Datos (CSV como ejemplo):**
    *   **Pandas:** `df_pd = pd.read_csv('data.csv')`
    *   **Polars:** `df_pl = pl.read_csv('data.csv')`
    (Polars también soporta `read_csv_batched` para lectura perezosa y eficiente).

4.  **Selección de Columnas:**
    *   **Pandas:** `df_pd[['col1', 'col2']]`
    *   **Polars:** `df_pl.select(['col1', 'col2'])` o `df_pl.select(pl.col('col1'), pl.col('col2'))`

5.  **Filtrado (where):**
    *   **Pandas:** `df_pd[df_pd['col1'] > 1]`
    *   **Polars:** `df_pl.filter(pl.col('col1') > 1)`

6.  **Creación de Nuevas Columnas (with_columns):**
    *   **Pandas:** `df_pd['new_col'] = df_pd['col1'] * 2`
    *   **Polars:** `df_pl.with_columns((pl.col('col1') * 2).alias('new_col'))`

7.  **Agregaciones (group_by):**
    *   **Pandas:** `df_pd.groupby('col1')['col2'].sum()`
    *   **Polars:** `df_pl.group_by('col1').agg(pl.col('col2').sum().alias('sum_col2'))`

8.  **Ordenamiento (sort):**
    *   **Pandas:** `df_pd.sort_values('col1')`
    *   **Polars:** `df_pl.sort('col1')`

9.  **Joins:**
    *   **Pandas:** `pd.merge(df1_pd, df2_pd, on='id', how='inner')`
    *   **Polars:** `df1_pl.join(df2_pl, on='id', how='inner')`

10. **Valores Nulos:**
    *   **Pandas:** `df_pd.dropna()`, `df_pd.fillna(0)`
    *   **Polars:** `df_pl.drop_nulls()`, `df_pl.fill_null(0)`
