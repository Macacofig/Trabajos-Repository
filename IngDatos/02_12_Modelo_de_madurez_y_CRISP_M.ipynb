# Modelo de madurez de análisis de datos de Gartner

El [modelo de madurez de análisis de datos de Gartner](https://computd.nl/4-levels-of-data-maturity/) es un marco que evalúa las capacidades analíticas de una organización en cuatro fases: descriptivo, diagnóstico, predictivo y prescriptivo. Este modelo ayuda a las empresas a comprender su nivel actual en el uso de datos para el soporte de la toma de decisiones, identificar áreas de mejora y planificar inversiones estratégicas para avanzar hacia una toma de decisiones más sofisticada y automatizada.

## Beneficios
- Permite a las organizaciones evaluar su situación actual en relación con las mejores prácticas de la industria.
- Ayuda a planificar inversiones en tecnologías y habilidades para mejorar los procesos analíticos.
- Facilita la identificación de brechas y oportunidades de mejora.
- Muestra el camino para pasar de una cultura de gestión de datos reactiva a una proactiva y orientada a la innovación.

## Niveles de madurez

### Nivel 1: Análisis Descriptivo
Se enfoca en entender "qué pasó". Las organizaciones utilizan los datos para informar sobre eventos pasados a través de informes y cuadros de mando básicos.

Sistemas operativos en producción.

### Nivel 2: Análisis de Diagnóstico:
Va un paso más allá y busca responder "¿por qué pasó?". Se investigan las causas subyacentes de los resultados, identificando relaciones y patrones en los datos.

DataMarts, Data Warehouse, ETLs, OLAP

### Nivel 3: Análisis Predictivo:
Estimar "qué pasará en el futuro". Se utilizan modelos para pronosticar resultados futuros basándose en datos históricos y análisis estadísticos.

Desarrollo, despliegue y actualización de modelos predictivos (AI y ML), CRISP-DM (Cross-Industry Standard Process for Data Mining)

### Nivel 4: Análisis Prescriptivo:
Recomienda "¿qué debemos hacer?". Este es el nivel más avanzado, donde se utilizan algoritmos y simulaciones para sugerir acciones óptimas y automatizar la toma de decisiones.

- Inteligencia Artificial (IA) y Machine Learning (ML)
- Modelos matemáticos complejos y las herramientas de simulación se usan para evaluar el impacto de diferentes escenarios de acción y determinar la mejor opción a seguir.
- BPM para integrar las recomendaciones prescriptivas en los flujos de trabajo existentes para automatizar la toma de decisiones y la ejecución de acciones.

# Cross-Industry Standard Process for Data Mining - CRISP-DM

Metodología estructurada y probada para la ejecución de proyectos de ciencia de datos. Proporciona un marco con seis fases (comprensión empresarial, comprensión de datos, preparación de datos, modelado, evaluación y despliegue).

CRISP-DM es flexible, permitiendo la iteración y el retroceso entre fases, y es ampliamente utilizado en diversas industrias para la toma de decisiones basada en datos.

- **Comprensión empresarial**: Define los objetivos del negocio y los requisitos del proyecto.
- **Comprensión de los datos**: Recopila los datos iniciales y explora la calidad y las características de los mismos.
- **Preparación de los datos**: Realiza tareas como la limpieza, transformación y selección de los datos para el modelado.
- **Modelado**: Aplica diversas técnicas de modelado para buscar patrones y relaciones.
- **Evaluación**: Evalúa los modelos para determinar si cumplen con los objetivos empresariales y decidir el próximo paso.
- **Despliegue**: Implementa los modelos en el entorno de negocio y realiza el seguimiento y mantenimiento.



# Sistemas basados en reglas vs Aprendizaje automático

Los sistemas basados en reglas utilizan reglas predefinidas de expertos humanos para tomar decisiones, mientras que el aprendizaje automático (ML) aprende patrones directamente de los datos. Los sistemas de reglas son rígidos, pero predecibles y transparentes, y funcionan bien para problemas bien definidos, aunque requieren intervención humana para cualquier cambio o para escalar. Por el contrario, los sistemas de ML son adaptables, pueden manejar datos complejos y no estructurados, y mejoran con el tiempo sin programación explícita, pero requieren grandes cantidades de datos para entrenarse.

## Sistemas basados en reglas

**Funcionamiento**: Un conjunto de reglas (a menudo con lógica "si/entonces") codificadas por expertos humanos toma decisiones. Por ejemplo, "Si la temperatura es inferior a 5 grados y hay lluvia, entonces activa el sistema de calefacción".

Ventajas:
- Fáciles de entender y explicar la lógica.
- Eficaces para tareas con reglas claras y predecibles.
- Requieren menos datos de entrenamiento que el ML.

Desventajas:
- Rígidos y poco adaptables a situaciones nuevas o imprevistas.
- Difíciles de mantener y escalar a medida que el número de reglas aumenta.
- Dependen completamente del conocimiento experto humano.

## Aprendizaje automático

**Funcionamiento**: Los algoritmos analizan grandes conjuntos de datos históricos para identificar patrones, crear modelos predictivos y mejorar su precisión de forma continua sin programación explícita.

Ventajas:
- Altamente adaptables y dinámicos.
- Pueden manejar datos complejos y no estructurados.
- A medida que se les alimenta con más datos, su rendimiento mejora automáticamente.

Desventajas:
- Requieren [grandes volúmenes de datos de entrenamiento de calidad](https://www.iic.uam.es/innovacion/inteligencia-artificial-centrada-en-datos-calidad-vs-cantidad/).
- La lógica puede ser opaca ("caja negra"), lo que dificulta explicar por qué se tomó una decisión específica.
- Pueden heredar sesgos de los datos de entrenamiento.
