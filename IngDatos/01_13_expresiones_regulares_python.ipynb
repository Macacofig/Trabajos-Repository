## Grupos sin captura (Non-capturing groups)

Los grupos sin captura te permiten agrupar partes de una expresión regular sin crear un grupo de captura numerado o nombrado. Esto es útil cuando necesitas agrupar elementos para aplicar cuantificadores o alternaciones, pero no necesitas acceder al contenido de ese grupo por separado.

Se definen utilizando `(?:...)`.

### Definición

Un grupo de captura normal `(...)` crea un grupo al que puedes referenciar posteriormente (por ejemplo, usando `\1` en la expresión regular o accediendo a los grupos del objeto `match`). Un grupo sin captura `(?:...)` agrupa los elementos dentro de él para propósitos de aplicación de operadores (como cuantificadores `*`, `+`, `?`, `{m,n}` o alternación `|`), pero no "captura" el texto emparejado para su posterior recuperación.

### Ejemplos

Vamos a ver algunos ejemplos para ilustrar la diferencia.

````python
import re

texto_ejemplo = "La casa es roja o azul."

# Ejemplo con grupo de captura
patron_captura = r"La casa es (roja|azul)\."
match_captura = re.search(patron_captura, texto_ejemplo)

if match_captura:
    print("Con grupo de captura:")
    print(f"Texto completo emparejado: {match_captura.group(0)}")
    print(f"Contenido del grupo capturado: {match_captura.group(1)}")

print("-" * 20)

# Ejemplo con grupo sin captura
patron_no_captura = r"La casa es (?:roja|azul)\."
match_no_captura = re.search(patron_no_captura, texto_ejemplo)

if match_no_captura:
    print("Con grupo sin captura:")
    print(f"Texto completo emparejado: {match_no_captura.group(0)}")
    try:
        # Intentar acceder a un grupo que no existe resultará en un error
        print(f"Contenido del grupo capturado: {match_no_captura.group(1)}")
    except IndexError as e:
        print(f"Error al intentar acceder a un grupo capturado (como se esperaba): {e}")
````

En el ejemplo anterior, con el grupo de captura `(roja|azul)`, podemos acceder al color emparejado usando `match_captura.group(1)`. Con el grupo sin captura `(?:roja|azul)`, la alternación funciona para emparejar "roja" o "azul", pero no se crea un grupo de captura, por lo que intentar acceder a `match_no_captura.group(1)` genera un `IndexError`.

Otro ejemplo con cuantificadores:

````python
import re

texto_cuantificador = "ababab"

# Ejemplo con grupo de captura y cuantificador
patron_captura_cuantificador = r"(ab)+"
match_captura_cuantificador = re.search(patron_captura_cuantificador, texto_cuantificador)

if match_captura_cuantificador:
    print("Con grupo de captura y cuantificador:")
    print(f"Texto completo emparejado: {match_captura_cuantificador.group(0)}")
    # Nota: Con cuantificadores, el grupo de captura solo guarda el ÚLTIMO emparejamiento
    print(f"Contenido del grupo capturado: {match_captura_cuantificador.group(1)}")

print("-" * 20)

# Ejemplo con grupo sin captura y cuantificador
patron_no_captura_cuantificador = r"(?:ab)+"
match_no_captura_cuantificador = re.search(patron_no_captura_cuantificador, texto_cuantificador)

if match_no_captura_cuantificador:
    print("Con grupo sin captura y cuantificador:")
    print(f"Texto completo emparejado: {match_no_captura_cuantificador.group(0)}")
    try:
        print(f"Contenido del grupo capturado: {match_no_captura_cuantificador.group(1)}")
    except IndexError as e:
        print(f"Error al intentar acceder a un grupo capturado (como se esperaba): {e}")
````

En este segundo ejemplo, ambos patrones `(ab)+` y `(?:ab)+` emparejan la cadena completa "ababab". Sin embargo, con el grupo de captura `(ab)+`, `group(1)` solo contiene el último "ab" emparejado por el grupo. Con el grupo sin captura `(?:ab)+`, no hay un grupo capturado al que acceder.

### Casos de uso y buenas prácticas

- **Agrupación para operadores:** Utiliza grupos sin captura cuando necesites agrupar partes de tu patrón para aplicar cuantificadores (`*`, `+`, `?`, `{m,n}`) o alternaciones (`|`), pero no necesites extraer el contenido de ese grupo.
- **Rendimiento:** En algunos motores de expresiones regulares, los grupos sin captura pueden ser ligeramente más eficientes ya que no necesitan almacenar el contenido emparejado. Aunque la diferencia puede ser mínima en la mayoría de los casos, es una buena práctica utilizarlos cuando la captura no es necesaria.
- **Simplificar la indexación de grupos:** Al no crear grupos de captura adicionales, los grupos sin captura pueden hacer que tus referencias a grupos de captura numerados sean más predecibles. Si tienes una expresión regular compleja con muchos grupos, el uso de grupos sin captura donde no necesitas la captura evita que tengas que recalcular constantemente los índices de tus grupos capturados.
- **Claridad:** Usar `(?:...)` comunica la intención de que esa parte del patrón es para agrupamiento lógico y no para extracción de datos.

En resumen, utiliza grupos sin captura cuando necesites agrupar partes de tu expresión regular para controlar la aplicación de operadores, pero no necesites acceder al texto que emparejan esos grupos. Esto puede mejorar ligeramente el rendimiento y la claridad de tu patrón.
