La función `dict()` en Python es muy versátil para crear diccionarios. Aquí te explico las formas más comunes:

### 1. Diccionario vacío

Puedes crear un diccionario vacío simplemente llamando a `dict()` sin argumentos.

````python
vacio = dict()
print(vacio)
````

### 2. A partir de argumentos de palabras clave (keyword arguments)

Esta es una forma concisa de crear un diccionario si ya conoces las claves y los valores. Las claves se pasan como argumentos con nombre.

````python
diccionario_kwargs = dict(nombre='Alice', edad=30, ciudad='Nueva York')
print(diccionario_kwargs)
````

### 3. A partir de un iterable de pares clave-valor

`dict()` puede tomar un iterable (como una lista de tuplas o listas) donde cada elemento es un par `(clave, valor)`.

````python
lista_de_pares = [('fruta', 'manzana'), ('color', 'rojo'), ('cantidad', 5)]
diccionario_desde_lista = dict(lista_de_pares)
print(diccionario_desde_lista)

# También funciona con tuplas
tuplas_de_pares = (('país', 'Bolivia'), ('continente', 'Sudamérica'))
diccionario_desde_tuplas = dict(tuplas_de_pares)
print(diccionario_desde_tuplas)
````

### 4. A partir de otro diccionario (copia)

Si pasas otro diccionario a `dict()`, creará una copia superficial de ese diccionario.

````python
diccionario_original = {'a': 1, 'b': 2}
diccionario_copia = dict(diccionario_original)
print(diccionario_copia)
````

### 5. Combinando `zip()` con `dict()`

Una forma muy común y útil es combinar `dict()` con la función `zip()`, que empareja elementos de dos o más iterables. Esto es ideal cuando tienes una lista de claves y una lista de valores por separado.

````python
claves = ['uno', 'dos', 'tres']
valores = [1, 2, 3]
diccionario_zip = dict(zip(claves, valores))
print(diccionario_zip)
````

# Conjuntos

En Python los conjuntos tienen la siguiente características:

- A diferencia de las listas, no preservan el order en el que fueron agregados los elementos.
- No se permite elementos duplicados


````python
{1, 3, -1 , 3, 2}
````

````python
# 1. Conjunto vacío
conjunto_vacio = set()
print(f"Conjunto vacío: {conjunto_vacio}")
print(f"Tipo: {type(conjunto_vacio)}")

# 2. A partir de una lista
lista_numeros = [1, 2, 3, 2, 4, 1, 5]
conjunto_desde_lista = set(lista_numeros)
print(f"Desde lista: {conjunto_desde_lista}")

# 3. A partir de una tupla
tupla_letras = ('a', 'b', 'c', 'a', 'd')
conjunto_desde_tupla = set(tupla_letras)
print(f"Desde tupla: {conjunto_desde_tupla}")

# 4. A partir de una cadena
cadena_ejemplo = "abracadabra"
conjunto_desde_cadena = set(cadena_ejemplo)
print(f"Desde cadena: {conjunto_desde_cadena}")

# 5. A partir de otro conjunto (copia)
conjunto_original = {10, 20, 30}
conjunto_copia = set(conjunto_original)
print(f"Desde otro conjunto (copia): {conjunto_copia}")
````

# Duck typing en Python

### ¿Qué es Duck Typing en Python?

"Duck typing" (tipado de pato) es un estilo de tipado que se usa en lenguajes de programación dinámicos como Python. Se refiere al principio de que el tipo de un objeto se determina por **lo que puede hacer** (sus métodos y propiedades), no por su herencia o implementación explícita.

La frase que resume este concepto es: "Si camina como un pato y grazna como un pato, entonces es un pato" (If it walks like a duck and quacks like a duck, then it must be a duck).

En Python, esto significa que no necesitas declarar explícitamente que un objeto pertenece a una clase particular para que sea tratado como tal. En su lugar, si un objeto tiene los métodos o atributos necesarios para una operación, entonces esa operación se puede realizar con ese objeto.

**Ventajas:**
*   **Flexibilidad:** Permite escribir código más genérico y reusable.
*   **Menos acoplamiento:** El código no depende de tipos concretos, sino de la "interfaz" que el objeto presenta.
*   **Facilita el polimorfismo:** Diferentes objetos pueden responder a la misma llamada de método, siempre y cuando lo implementen, sin necesidad de una jerarquía de clases común explícita.

**Este concepto en MUY importante en Python** porque hace posible que, por ejemplo, un objeto que NO es intrínsecamente una lista, soporte y se comporte como si lo fuera (es decir, implementa los métodos o el protocolo necesario para las operaciones de lista, como la indexación o la iteración).

Cuando decimos que un objeto "soporta las operaciones de una lista", lo que realmente significa es que ese objeto implementa los protocolos necesarios para esas operaciones. Por ejemplo, para ser iterable, un objeto debe implementar el protocolo iterable (generalmente con el método `__iter__`). Para ser un contenedor que soporte la verificación de pertenencia (in), debe implementar el protocolo de contenedor (con el método `__contains__`).

Entonces, el Duck Typing observa si un objeto cumple con el protocolo requerido (es decir, tiene los métodos y atributos correctos), sin importar su tipo formal.

Para facilitar su uso muchas librerías tienen objetos que se comportan como algun tipo de iterable de esta manera si sabemos como usar los iterables es más fácil usar la librería.
