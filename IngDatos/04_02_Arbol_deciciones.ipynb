## Arboles de decisión

Los **árboles de decisión** son uno de los algoritmos más intuitivos y populares en el **aprendizaje automático supervisado**. Se utilizan tanto para **clasificación** como para **regresión**, y su funcionamiento se basa en dividir los datos en subconjuntos según las características que mejor separan las clases o explican la variable objetivo.

Un árbol de decisión es un clasificador que toma como entrada una entidad/ejemplo descrita por un conjunto de atributos y devuelve una «decisión». Los atributos de entrada pueden ser discretos o continuos.

Un árbol de decisión aplica una secuencia de tests para poder alcanzar la decisión. Cada nodo interno del árbol corresponde con un test sobre el valor de uno de los atributos, y las ramas que salen del nodo están etiquetadas con los posibles valores de dicho atributos. Cada nodo hoja del árbol representa el valor que ha de ser devuelto si dicho nodo hoja es alcanzado.

La representación en forma de árboles de decisión es muy natural para los humanos; muchos manuales que explican cómo hacer determinadas tareas (por ejemplo, reparar un coche) están escritos, en general, como un árbol de decisión.

La idea básica del algoritmo APRENDIZAJE-ÁRBOL-DECISIÓN es realizar primero el test sobre el atributo más importante. Se considera como atributo «más importante»: aquel que clasifica la mayor cantidad de ejemplos en una clase/categoría. De esta forma, esperamos obtener la clasificación correcta con el menor número posible de tests; es decir, con un árbol en el que las ramas sean cortas.

```python
#Nota: log(0) no está definido
import math

def cantidad_info(*ps):
    return sum([-p * (0 if p == 0 else math.log2(p)) for p in ps])
```

```python
#Cuando no sabemos nada sobre la moneda, necesitamos 1 bit de información para responder la pregunta sobre si saldrá cara o cruz
cantidad_info(1/2, 1/2)
```

```python
#Cuando sabemos que la moneda está cargada para que salga cara 9 de cada 10 veces,
#NO necesitamos 1 bit de información para responder la pregunta sobre si saldrá cara o cruz
# cantidad_info(9/10, 1/10)
# cantidad_info(99/100, 1/100)
```

## Ganancia de información

image.png




image.png

```python
# si = 6/12, no = 6/12
cantidad_de_info_inicial = cantidad_info(6/12, 6/12)
cantidad_de_info_inicial
```

```python
#PRECIO: $$$, $$, $
info_precio = 3/12*cantidad_info(1/3, 2/3) + 2/12*cantidad_info(2/2, 0/2) + 7/12*cantidad_info(3/7,4/7)
info_precio
```

```python
#Ganacia precio
cantidad_de_info_inicial - info_precio
```

```python
#TIPO DE RESTAURANTE: francés 2 casos;italiano 2 casos;tai 4 casos;hamburguesería 4 casos
info_tipo = 2/12*cantidad_info(1/2, 1/2) + 2/12*cantidad_info(1/2, 1/2) + 4/12*cantidad_info(2/4,2/4)  + 4/12*cantidad_info(2/4,2/4)
info_tipo
```

```python
#Ganacia tipo
cantidad_de_info_inicial - info_tipo
```

```python
##CANTIDAD DE CLIENTES: ninguno 2 casos;algunos 4 casos;lleno 6 casos
info_clientes = 2/12*cantidad_info(0, 1) + 4/12*cantidad_info(1, 0) + 6/12*cantidad_info(2/6,4/6)
info_clientes
```

```python
#Ganacia número de clientes
cantidad_de_info_inicial - info_clientes
```

```python
#Calcule la ganacia de información para la variable 'hambriento'
info_hambriento = 7/12*cantidad_info(5/7, 2/7) + 5/12*cantidad_info(1/5, 4/5)
info_hambriento
```

```python
#Ganacia hambriento
cantidad_de_info_inicial - info_hambriento
```
