```python
html_inventory = """
<!DOCTYPE html>
<html>
  <body>
    <div class="inventario">
        <h1>Lista de Gadgets 2026</h1>

        <div class="producto" id="p1">
            <span class="categoria">Smartphone</span>
            <h2 class="nombre">iPhone 17 Pro</h2>
            <p class="estado">En Stock</p>
            <span class="precio">$999</span>
        </div>

        <div class="producto" id="p2">
            <span class="categoria">Laptop</span>
            <h2 class="nombre">MacBook Pro M5</h2>
            <p class="estado">Agotado</p>
            <span class="precio">$2499</span>
        </div>

        <div class="producto" id="p3">
            <span class="categoria">Tablet</span>
            <h2 class="nombre">iPad Ultra</h2>
            <p class="estado">Preventa</p>
            <span class="precio">$1299</span>
        </div>
    </div>
  </body>
</html>
"""
```

```python
from bs4 import BeautifulSoup
page = BeautifulSoup(html_inventory, 'html.parser')
# page.body
```

```python
h2_nombre_productos = page.body.find_all("h2", attrs={"class":"nombre"})
```

```python
h2_iphone = [h2_nombre_producto
 for h2_nombre_producto in h2_nombre_productos
 if h2_nombre_producto.text == 'iPhone 17 Pro'][0]
```

```python
h2_iphone
```

```python
iphone_div = h2_iphone.find_parent()
iphone_div.find('span', attrs={"class":"precio"}).text
```
