## Principio de Única Responsabilidad (SRP - Single Responsibility Principle)

El Principio de Única Responsabilidad es uno de los cinco principios SOLID de diseño de software. Fue introducido por Robert C. Martin (Uncle Bob).

### Definición

Este principio establece que **una clase o módulo debe tener solo una razón para cambiar**, o, en otras palabras, **una clase debe tener una y solo una responsabilidad**. Una responsabilidad se puede entender como una razón para que una entidad (clase, función, módulo) cambie. Si una clase tiene más de una razón para cambiar, entonces tiene múltiples responsabilidades, y esto va en contra del SRP.

Es importante destacar que el SRP no se limita solo a clases; puede aplicarse a funciones, módulos o incluso microservicios.

### Ventajas de Aplicar el SRP

Aplicar el Principio de Única Responsabilidad en el diseño de software trae consigo múltiples beneficios:

1.  **Mayor Cohesión**: Las clases y funciones se vuelven más cohesivas, ya que todos sus elementos están relacionados con una única responsabilidad. Esto significa que las partes de una clase trabajan juntas para un propósito bien definido.
2.  **Menor Acoplamiento**: Al tener una única responsabilidad, las clases dependen menos de otras clases para realizar sus tareas. Esto reduce el acoplamiento entre los componentes del sistema, lo que facilita el mantenimiento y la evolución.
3.  **Facilidad de Mantenimiento**: Cuando una clase solo tiene una razón para cambiar, los cambios en una parte del sistema tienen menos probabilidades de afectar otras partes. Esto simplifica la depuración y la modificación del código.
4.  **Mayor Claridad y Legibilidad**: Las clases con una única responsabilidad son más fáciles de entender. El propósito de la clase es claro, y su comportamiento es predecible.
5.  **Facilidad de Prueba**: Las clases pequeñas y enfocadas en una única tarea son mucho más sencillas de probar de forma unitaria, ya que tienen menos dependencias y casos de uso complejos.
6.  **Reutilización del Código**: Componentes bien definidos y con una única responsabilidad son más fáciles de reutilizar en diferentes partes del sistema o incluso en otros proyectos.
7.  **Mejor Escalabilidad**: En sistemas grandes o distribuidos, una buena aplicación del SRP puede conducir a servicios más pequeños y especializados que son más fáciles de escalar de forma independiente.

### Ejemplos

Veamos un ejemplo clásico de cómo aplicar (o no aplicar) el SRP.

#### Ejemplo 1: Violación del SRP (Antipatrón)

Consideremos una clase `Reporte` que se encarga de generar un reporte, formatearlo y enviarlo por correo electrónico. Esta clase tiene **múltiples responsabilidades**:

*   **Generar datos del reporte**
*   **Formatear el reporte** (ej. a PDF, HTML)
*   **Enviar el reporte** (ej. por email)

Si la lógica para generar los datos cambia, la clase `Reporte` cambia. Si el formato del reporte cambia (ej. de PDF a Excel), la clase `Reporte` cambia. Si el método de envío cambia (ej. de email a FTP), la clase `Reporte` también cambia. ¡Tres razones para cambiar!

````python
class Reporte:
    def __init__(self, datos):
        self.datos = datos

    def generar_contenido(self):
        # Lógica para procesar los datos y generar el contenido del reporte
        print("Generando contenido del reporte...")
        return f"Contenido del reporte con datos: {self.datos}"

    def formatear_pdf(self, contenido):
        # Lógica para formatear el contenido como PDF
        print("Formateando reporte a PDF...")
        return f"<PDF>{contenido}</PDF>"

    def enviar_por_email(self, reporte_formateado, destinatario):
        # Lógica para enviar el reporte por email
        print(f"Enviando reporte a {destinatario}...")
        print(f"Reporte enviado: {reporte_formateado}")


# Uso de la clase que viola el SRP
reporte_diario = Reporte("Ventas del día")
contenido = reporte_diario.generar_contenido()
reporte_pdf = reporte_diario.formatear_pdf(contenido)
reporte_diario.enviar_por_email(reporte_pdf, "gerencia@empresa.com")
````

#### Ejemplo 2: Aplicando el SRP (Buenas Prácticas)

Para aplicar el SRP, debemos dividir la clase `Reporte` en varias clases, cada una con una única responsabilidad:

*   `GeneradorDatosReporte`: Responsable de obtener y procesar los datos para el reporte.
*   `FormateadorReporte`: Responsable de tomar los datos y darles un formato específico (PDF, HTML, etc.).
*   `EnviadorReporte`: Responsable de tomar un reporte formateado y enviarlo por el medio adecuado (email, FTP, etc.).

Ahora, si el formato del reporte cambia, solo se modifica `FormateadorReporte`. Si el método de envío cambia, solo se modifica `EnviadorReporte`.

````python
class GeneradorDatosReporte:
    def __init__(self, datos):
        self.datos = datos

    def obtener_datos(self):
        # Lógica para obtener y procesar los datos del reporte
        print("Obteniendo datos para el reporte...")
        return {"titulo": "Reporte Diario", "contenido": f"Contenido generado con: {self.datos}"}

class FormateadorReporte:
    def formatear_como_pdf(self, datos_reporte):
        # Lógica para formatear los datos como PDF
        print("Formateando reporte a PDF...")
        return f"<PDF>{datos_reporte['titulo']}\n{datos_reporte['contenido']}</PDF>"

    def formatear_como_html(self, datos_reporte):
        # Lógica para formatear los datos como HTML
        print("Formateando reporte a HTML...")
        return f"<html><body><h1>{datos_reporte['titulo']}</h1><p>{datos_reporte['contenido']}</p></body></html>"

class EnviadorReporte:
    def enviar_por_email(self, reporte_formateado, destinatario):
        # Lógica para enviar el reporte por email
        print(f"Enviando reporte por email a {destinatario}...")
        print(f"Reporte enviado: {reporte_formateado[:50]}...") # Muestra solo el inicio

    def enviar_por_ftp(self, reporte_formateado, url_ftp):
        # Lógica para enviar el reporte por FTP
        print(f"Subiendo reporte por FTP a {url_ftp}...")
        print(f"Reporte subido: {reporte_formateado[:50]}...") # Muestra solo el inicio


# Uso de las clases que aplican el SRP

# 1. Generar los datos
generador = GeneradorDatosReporte("Ventas del Mes")
datos_listos = generador.obtener_datos()

# 2. Formatear el reporte
formateador = FormateadorReporte()
reporte_final_pdf = formateador.formatear_como_pdf(datos_listos)
# reporte_final_html = formateador.formatear_como_html(datos_listos) # Podríamos usar otro formato

# 3. Enviar el reporte
enviador = EnviadorReporte()
enviador.enviar_por_email(reporte_final_pdf, "direccion@empresa.com")
# enviador.enviar_por_ftp(reporte_final_html, "ftp.empresa.com/reportes/") # O usar otro método de envío
````

Este ejemplo demuestra cómo al dividir las responsabilidades, logramos un código más modular, flexible y fácil de mantener, que es la esencia del Principio de Única Responsabilidad. ¡Espero que esto aclare sus dudas!

## Shell Imperativo, Core Funcional

El principio de 'Shell Imperativo, Core Funcional' (también conocido como 'Functional Core, Imperative Shell') es un patrón de diseño que aboga por separar la lógica de negocio pura, que es idealmente funcional y sin efectos secundarios, de las operaciones imperativas de entrada/salida (I/O) y mutación de estado. Fue popularizado por Gary Bernhardt en su charla "Boundaries".

### Definición

*   **Core Funcional**: Es la parte central de su aplicación que contiene la lógica de negocio más importante. Este "core" debe estar compuesto por funciones puras: para las mismas entradas, siempre producen las mismas salidas y no tienen efectos secundarios (no modifican el estado fuera de su ámbito ni realizan I/O).
*   **Shell Imperativo**: Es la capa externa de la aplicación que se encarga de interactuar con el mundo exterior. Aquí es donde ocurren todas las operaciones con efectos secundarios, como leer de una base de datos o un archivo, escribir en la consola, enviar un correo electrónico, interactuar con la red, etc. El shell llama a las funciones puras del core, pasa sus resultados al mundo exterior o toma entradas del exterior para pasarlas al core.

### Ventajas de Aplicar el 'Shell Imperativo, Core Funcional'

La adopción de este patrón de diseño ofrece varias ventajas significativas:

1.  **Facilidad de Prueba**: El core funcional, al ser puro y sin efectos secundarios, es extremadamente fácil de probar. No necesita mocks ni configuraciones complejas; simplemente se le pasan entradas y se verifica la salida. Esto reduce drásticamente la complejidad y el tiempo de las pruebas unitarias.
2.  **Mayor Confiabilidad**: Al aislar los efectos secundarios en el shell, el core se vuelve más predecible y menos propenso a errores. Los errores relacionados con el estado o I/O se contienen en el shell, facilitando su depuración.
3.  **Reutilización del Código**: Las funciones puras del core son altamente reutilizables. No dependen del contexto de I/O, por lo que pueden ser utilizadas en diferentes partes de la aplicación o incluso en otros proyectos con facilidad.
4.  **Claridad y Legibilidad**: La separación clara entre la lógica de negocio y las operaciones de I/O hace que el código sea más fácil de entender. Se sabe exactamente dónde buscar la lógica de negocio pura y dónde se manejan las interacciones externas.
5.  **Mantenimiento Simplificado**: Los cambios en las operaciones de I/O (por ejemplo, cambiar de un archivo a una base de datos) solo afectan al shell imperativo, dejando el core funcional intacto. De manera similar, los cambios en la lógica de negocio se pueden realizar en el core sin preocuparse por los efectos secundarios en el shell.
6.  **Programación Paralela y Concurrente**: Las funciones puras son inherentemente seguras para la concurrencia, ya que no comparten ni modifican el estado mutable. Esto facilita la construcción de sistemas paralelos y concurrentes.

### Ejemplo: Procesamiento de Pedidos

Consideremos un sistema de procesamiento de pedidos. La lógica de calcular el total de un pedido (core funcional) es independiente de cómo se recibe el pedido o cómo se guarda la factura (shell imperativo).

#### Core Funcional: Lógica de Cálculo Pura

````python
def calcular_total_pedido(items, tasas_impuesto):
    """Calcula el total de un pedido, incluyendo impuestos. Función pura."""
    subtotal = sum(item['cantidad'] * item['precio_unitario'] for item in items)
    total_impuestos = subtotal * tasas_impuesto['general']
    return subtotal + total_impuestos

def aplicar_descuento(total, porcentaje_descuento):
    """Aplica un descuento al total. Función pura."""
    return total * (1 - porcentaje_descuento / 100)

# Prueba del Core (fácil y sin efectos secundarios)
items_ejemplo = [
    {'producto': 'Laptop', 'cantidad': 1, 'precio_unitario': 1200},
    {'producto': 'Mouse', 'cantidad': 2, 'precio_unitario': 25}
]
tasas_impuesto_ejemplo = {'general': 0.10}

total_bruto = calcular_total_pedido(items_ejemplo, tasas_impuesto_ejemplo)
print(f"Total bruto calculado: {total_bruto}")

total_con_descuento = aplicar_descuento(total_bruto, 5)
print(f"Total con 5% de descuento: {total_con_descuento}")
````

#### Shell Imperativo: Manejo de I/O y Coordinación

Aquí, las funciones se encargarán de leer el pedido de una fuente (ej. JSON, base de datos) y de guardar la factura o enviar una confirmación.

````python
import json

def leer_pedido_desde_json(ruta_archivo):
    """Lee los detalles de un pedido desde un archivo JSON. Imperativo."""
    print(f"[SHELL] Leyendo pedido desde {ruta_archivo}...")
    try:
        with open(ruta_archivo, 'r') as f:
            data = json.load(f)
            return data['items'], data['tasas_impuesto'], data.get('descuento_porcentaje', 0)
    except FileNotFoundError:
        print(f"[SHELL] Error: Archivo {ruta_archivo} no encontrado.")
        return [], {}, 0
    except json.JSONDecodeError:
        print(f"[SHELL] Error: Formato JSON inválido en {ruta_archivo}.")
        return [], {}, 0

def guardar_factura_en_archivo(numero_pedido, total_final, ruta_salida):
    """Guarda la factura en un archivo de texto. Imperativo."""
    print(f"[SHELL] Guardando factura para el pedido {numero_pedido} en {ruta_salida}...")
    with open(ruta_salida, 'w') as f:
        f.write(f"Factura Pedido #{numero_pedido}\n")
        f.write(f"Total: {total_final:.2f} EUR\n")
        f.write("Gracias por su compra!")

def procesar_pedido(numero_pedido, ruta_entrada_json, ruta_salida_factura):
    """Función principal del shell que orquesta el proceso. Imperativa."""
    print(f"[SHELL] Iniciando procesamiento para el pedido #{numero_pedido}...")

    # 1. Entrada (Imperativo)
    items, tasas_impuesto, descuento_porcentaje = leer_pedido_desde_json(ruta_entrada_json)
    if not items:
        print(f"[SHELL] No se pudo procesar el pedido #{numero_pedido}.")
        return

    # 2. Lógica de Negocio (Core Funcional)
    total_bruto = calcular_total_pedido(items, tasas_impuesto)
    total_final = aplicar_descuento(total_bruto, descuento_porcentaje)

    # 3. Salida (Imperativo)
    guardar_factura_en_archivo(numero_pedido, total_final, ruta_salida_factura)
    print(f"[SHELL] Pedido #{numero_pedido} procesado con éxito. Total final: {total_final:.2f}")

# Simulación de un archivo de pedido
pedido_json_data = {
    "items": [
        {"producto": "Libro A", "cantidad": 1, "precio_unitario": 30},
        {"producto": "Libro B", "cantidad": 2, "precio_unitario": 20}
    ],
    "tasas_impuesto": {"general": 0.08},
    "descuento_porcentaje": 10
}

with open('pedido_123.json', 'w') as f:
    json.dump(pedido_json_data, f)

# Ejecución del Shell
procesar_pedido('123', 'pedido_123.json', 'factura_123.txt')
````

En este ejemplo, `calcular_total_pedido` y `aplicar_descuento` son funciones puras (core funcional), mientras que `leer_pedido_desde_json`, `guardar_factura_en_archivo` y `procesar_pedido` son parte del shell imperativo, manejando la interacción con archivos y coordinando las llamadas al core. Esto permite que la lógica central de cálculo sea aislada, fácil de probar y reutilizable.

## Diseño Top-Down (De Arriba Hacia Abajo)

El Diseño Top-Down es una estrategia de descomposición de sistemas que comienza con una vista de alto nivel del sistema y, progresivamente, desglosa este sistema en componentes más pequeños y manejables, hasta alcanzar un nivel de detalle donde cada componente es fácil de implementar y entender.

### Definición

En el Diseño Top-Down, se empieza por identificar la función principal o el objetivo general del sistema. Luego, este objetivo se divide en una serie de subfunciones o módulos principales. Cada uno de estos subobjetivos se descompone a su vez en tareas más pequeñas, y así sucesivamente, creando una jerarquía. El proceso continúa hasta que cada "hoja" de esta jerarquía representa una tarea lo suficientemente simple como para ser implementada individualmente. Este enfoque prioriza el entendimiento de la estructura general y las interacciones antes de sumergirse en los detalles de implementación.

Pueden imaginarlo como crear un índice para un libro grande antes de escribir los capítulos; primero los títulos principales, luego los subcapítulos, y así hasta llegar a los párrafos individuales.

### Ventajas del Diseño Top-Down

La aplicación del Diseño Top-Down ofrece varias ventajas significativas en el desarrollo de software:

1.  **Claridad Estructural**: Proporciona una visión clara y jerárquica de la arquitectura del sistema desde el principio, facilitando la comprensión de cómo las partes encajan en el todo.
2.  **Gestión de la Complejidad**: Permite abordar problemas complejos descomponiéndolos en subproblemas más pequeños y manejables, reduciendo la carga cognitiva y haciendo el proceso de desarrollo más controlable.
3.  **Modularidad**: Fomenta la creación de módulos bien definidos con responsabilidades claras, lo que mejora la reusabilidad y el mantenimiento del código.
4.  **Detección Temprana de Errores**: Al definir las interfaces entre módulos en etapas tempranas, se pueden identificar y corregir posibles problemas de integración antes de que se invierta demasiado tiempo en la implementación de detalles.
5.  **Facilita el Trabajo en Equipo**: Diferentes equipos o individuos pueden trabajar en diferentes submódulos en paralelo, una vez que las interfaces han sido definidas, lo que acelera el desarrollo.
6.  **Mayor Cohesión y Menor Acoplamiento**: Los módulos resultantes suelen tener alta cohesión (sus elementos están estrechamente relacionados con su función) y bajo acoplamiento (dependen poco de otros módulos), características deseables en el diseño de software.

### Ejemplo: Procesamiento de un Pedido Online (Diseño Top-Down)

Imaginemos que queremos desarrollar un sistema para procesar un pedido online. Usaremos el Diseño Top-Down para estructurar las funciones necesarias.

````python
def procesar_pedido_online(datos_pedido):
    """Función de alto nivel: Procesa un pedido online completo."""
    print("--- Iniciando procesamiento de pedido ---\n")

    # 1. Validar el pedido (función de nivel inferior)
    pedido_valido = validar_pedido(datos_pedido)
    if not pedido_valido:
        print("Pedido inválido. Finalizando.\n")
        return

    # 2. Calcular el total (función de nivel inferior)
    total = calcular_total_con_impuestos(datos_pedido['items'], datos_pedido['descuento'])

    # 3. Procesar el pago (función de nivel inferior)
    pago_exitoso = procesar_pago(total, datos_pedido['metodo_pago'])
    if not pago_exitoso:
        print("Error en el pago. Finalizando.\n")
        return

    # 4. Actualizar inventario (función de nivel inferior)
    inventario_actualizado = actualizar_inventario(datos_pedido['items'])
    if not inventario_actualizado:
        print("Error al actualizar inventario. Finalizando.\n")
        # Podríamos revertir el pago aquí en un sistema real
        return

    # 5. Enviar confirmación (función de nivel inferior)
    enviar_confirmacion_cliente(datos_pedido['email_cliente'], total)

    print(f"\n--- Pedido {datos_pedido.get('id_pedido', 'N/A')} procesado con éxito. Total: {total:.2f} ---")

def validar_pedido(pedido):
    """Valida la estructura y contenido básico del pedido."""
    print("  Validando pedido...")
    # Lógica de validación (ej: verificar que tenga items, cantidades válidas, etc.)
    if 'items' not in pedido or not pedido['items']:
        print("  - Error: El pedido no tiene items.")
        return False
    if 'email_cliente' not in pedido or '@' not in pedido['email_cliente']:
        print("  - Error: Email del cliente inválido.")
        return False
    print("  Pedido validado correctamente.")
    return True

def calcular_total_con_impuestos(items, descuento):
    """Calcula el total sumando items, aplicando impuestos y descuentos."""
    print("  Calculando total del pedido...")
    subtotal = sum(item['cantidad'] * item['precio_unitario'] for item in items)
    impuestos = subtotal * 0.10  # Suponemos 10% de impuesto
    total_bruto = subtotal + impuestos
    total_final = total_bruto * (1 - descuento / 100) # Aplica descuento
    print(f"  Total calculado: {total_final:.2f}")
    return total_final

def procesar_pago(monto, metodo):
    """Simula el procesamiento de un pago."""
    print(f"  Procesando pago de {monto:.2f} vía {metodo}...")
    # Lógica real de pasarela de pago
    if metodo == 'tarjeta' and monto > 0:
        print("  Pago con tarjeta exitoso.")
        return True
    elif metodo == 'paypal' and monto > 0:
        print("  Pago con PayPal exitoso.")
        return True
    print("  Error: Método de pago no soportado o monto inválido.")
    return False

def actualizar_inventario(items):
    """Simula la actualización del inventario para cada item."""
    print("  Actualizando inventario...")
    # Lógica real de base de datos de inventario
    for item in items:
        print(f"  - Reduciendo stock de {item['producto']} en {item['cantidad']}.")
    print("  Inventario actualizado correctamente.")
    return True

def enviar_confirmacion_cliente(email, total):
    """Simula el envío de un correo electrónico de confirmación."""
    print(f"  Enviando confirmación a {email} con total {total:.2f}...")
    # Lógica real de envío de email
    print("  Confirmación enviada.")
    return True

# --- Uso del sistema con Diseño Top-Down ---

# Definición del pedido a un alto nivel
pedido_ejemplo = {
    'id_pedido': 'ABC-123',
    'items': [
        {'producto': 'Teclado Mecánico', 'cantidad': 1, 'precio_unitario': 100},
        {'producto': 'Mouse Gaming', 'cantidad': 1, 'precio_unitario': 50}
    ],
    'descuento': 10, # 10% de descuento
    'metodo_pago': 'tarjeta',
    'email_cliente': 'cliente@ejemplo.com'
}

procesar_pedido_online(pedido_ejemplo)

print("\n--- Probando con un pedido inválido ---")
pedido_invalido = {
    'id_pedido': 'XYZ-456',
    'items': [], # Sin items, esto hará que la validación falle
    'descuento': 0,
    'metodo_pago': 'paypal',
    'email_cliente': 'otro@ejemplo.com'
}
procesar_pedido_online(pedido_invalido)
````

En este ejemplo, la función `procesar_pedido_online` es el nivel más alto. Esta función se encarga de orquestar el flujo general del procesamiento, pero delega las tareas específicas (validar, calcular, pagar, actualizar, confirmar) a funciones de nivel inferior. Cada una de estas funciones de nivel inferior podría, a su vez, dividirse en funciones aún más pequeñas si su complejidad lo justificara. Este enfoque nos permite entender el flujo general antes de preocuparnos por los detalles de cómo se implementa cada paso, y demuestra claramente las ventajas de modularidad y claridad del Diseño Top-Down.

## Diseño por Contrato (Design by Contract - DbC)

El **Diseño por Contrato (DbC)** es una metodología de diseño de software que se basa en la metáfora de las obligaciones y beneficios mutuos en una relación de negocios entre un "cliente" (quien llama una operación) y un "proveedor" (quien implementa la operación). Fue propuesto por Bertrand Meyer en el contexto del lenguaje de programación Eiffel.

En DbC, las interacciones entre los componentes del software se formalizan a través de "contratos", que son conjuntos de aserciones:

*   **Precondiciones**: Requisitos que deben ser verdaderos antes de que una operación (función o método) sea llamada.
*   **Postcondiciones**: Garantías que el proveedor promete cumplir al finalizar la operación, siempre que las precondiciones se hayan satisfecho.
*   **Invariantes**: Condiciones que deben ser verdaderas antes y después de cualquier operación pública de una clase o módulo.

### Usos y Ventajas del DbC

*   **Mejora la calidad del software**: Ayuda a prevenir errores al detectarlos temprano en el ciclo de desarrollo.
*   **Documentación viva**: Los contratos actúan como una forma de documentación ejecutables que aclaran la intención de los desarrolladores.
*   **Depuración más fácil**: Cuando un contrato se rompe, se señala la fuente del error, lo que facilita la depuración.
*   **Modularidad**: Fomenta el diseño de módulos con interfaces claras y responsabilidades bien definidas.
*   **Reutilización y extensibilidad**: Los componentes con contratos claros son más fáciles de entender, reutilizar y extender.

Veamos cada uno en detalle.

### Precondiciones (Preconditions)

Las **precondiciones** son condiciones que deben ser verdaderas justo antes de que una función o método sea ejecutado. Son responsabilidades del **cliente** (quien llama a la función). Si el cliente no cumple con las precondiciones, el comportamiento del proveedor no está garantizado.

**Usos:**
*   Validar la entrada de parámetros para asegurar que son apropiados para la operación.
*   Asegurar que el estado del objeto o sistema es válido antes de realizar una acción.
*   Prevenir divisiones por cero, accesos a índices fuera de rango, o el uso de objetos nulos.

**Ejemplo:**

````python
def dividir(numerador, denominador):
    # Precondición: el denominador no debe ser cero
    if denominador == 0:
        raise ValueError("El denominador no puede ser cero")

    return numerador / denominador

# Uso correcto (cumple la precondición)
print(f"Resultado de la división: {dividir(10, 2)}")

# Uso incorrecto (viola la precondición)
try:
    print(dividir(10, 0))
except ValueError as e:
    print(f"Error al dividir: {e}")
````

### Postcondiciones (Postconditions)

Las **postcondiciones** son condiciones que deben ser verdaderas justo después de que una función o método haya terminado su ejecución exitosamente. Son responsabilidades del **proveedor** (la función o método en sí).

**Usos:**
*   Verificar que el resultado de una función es correcto y cumple con las especificaciones.
*   Asegurar que el estado del objeto o sistema ha sido modificado de manera esperada.
*   Garantizar que los efectos secundarios de la operación son los previstos.

**Ejemplo:**

````python
def sumar_positivos(a, b):
    # Precondición: ambos números deben ser positivos
    if a <= 0 or b <= 0:
        raise ValueError("Ambos números deben ser positivos")

    resultado = a + b

    # Postcondición: el resultado debe ser mayor que cada uno de los sumandos
    if not (resultado > a and resultado > b):
        raise AssertionError("La suma de dos positivos debería ser mayor que sus componentes individuales")

    return resultado

# Uso correcto (cumple pre y postcondición)
print(f"Resultado de la suma: {sumar_positivos(5, 3)}")

# Ejemplo de cómo una postcondición podría detectar un error (hipotético)
def sumar_con_bug(a, b):
    if a <= 0 or b <= 0:
        raise ValueError("Ambos números deben ser positivos")

    resultado = a * b # ¡Error! Debería ser a + b

    if not (resultado > a and resultado > b):
        raise AssertionError("La suma de dos positivos debería ser mayor que sus componentes individuales")
    return resultado

try:
    sumar_con_bug(5, 3)
except AssertionError as e:
    print(f"Error detectado por postcondición: {e}")
````

### Invariantes (Invariants)

Los **invariantes** son condiciones que deben ser verdaderas en todo momento que un objeto esté en un estado "estable" (visible desde el exterior). Se deben cumplir antes y después de la ejecución de cualquier método público del objeto. Los invariantes son responsabilidades del **proveedor** (la clase o módulo).

**Usos:**
*   Mantener la consistencia interna de un objeto.
*   Garantizar la integridad de la estructura de datos que un objeto representa.
*   Reforzar las reglas de negocio que rigen el estado de un objeto.

**Ejemplo:**

````python
class CuentaBancaria:
    def __init__(self, saldo_inicial):
        # Invariante: El saldo nunca debe ser negativo
        if saldo_inicial < 0:
            raise ValueError("El saldo inicial no puede ser negativo")
        self._saldo = saldo_inicial
        self._verificar_invariante()

    def _verificar_invariante(self):
        # Método interno para verificar el invariante de la clase
        if self._saldo < 0:
            raise AssertionError("Invariante violado: El saldo de la cuenta no puede ser negativo")

    def depositar(self, monto):
        # Precondición: el monto a depositar debe ser positivo
        if monto <= 0:
            raise ValueError("El monto a depositar debe ser positivo")

        self._saldo += monto
        # Postcondición: El saldo debe haber aumentado en 'monto'
        # Y el invariante debe seguir siendo válido
        self._verificar_invariante()

    def retirar(self, monto):
        # Precondición: el monto a retirar debe ser positivo
        if monto <= 0:
            raise ValueError("El monto a retirar debe ser positivo")
        # Precondición: no se puede retirar más de lo que se tiene
        if self._saldo < monto:
            raise ValueError("Saldo insuficiente")

        self._saldo -= monto
        # Postcondición: El saldo debe haber disminuido en 'monto'
        # Y el invariante debe seguir siendo válido
        self._verificar_invariante()

    def obtener_saldo(self):
        self._verificar_invariante() # Verificar el invariante antes de devolver el estado
        return self._saldo


# --- Uso de la clase ---
cuenta = CuentaBancaria(100)
print(f"Saldo inicial: {cuenta.obtener_saldo()}")

cuenta.depositar(50)
print(f"Saldo después del depósito: {cuenta.obtener_saldo()}")

try:
    cuenta.retirar(200) # Saldo insuficiente, viola precondición
except ValueError as e:
    print(f"Error al retirar: {e}")

cuenta.retirar(30)
print(f"Saldo después del retiro: {cuenta.obtener_saldo()}")

try:
    # Si hubiera un error interno que dejara el saldo negativo, el invariante lo detectaría
    # Por ejemplo, una línea incorrecta como self._saldo = -10 dentro de un método
    pass # No hay tal bug aquí, solo para ilustrar
except AssertionError as e:
    print(f"Error detectado por invariante: {e}")
````

En este ejemplo de `CuentaBancaria`, el invariante asegura que el `_saldo` nunca es negativo. Cualquier método que modifique el saldo debe asegurar que este invariante se mantenga. Las precondiciones verifican las entradas antes de actuar, y las postcondiciones (implícitas aquí al verificar el invariante después de cada operación) aseguran que el estado final es consistente.

## Ejemplos de código con vicios/malos olores en el código

**Combinar cálculos con operaciones de entrada y salida en la misma función**


En general, es recomendable separar los cálculos (consultas) de las operaciones de entrada/salida (comandos) en la misma función (archivos, bases de datos, servicios web, consola)

```
User user = validar_usuario(id, password)
```

Si, por ejemplo, resolver un problema implica leer desde un archivo, manipular los contenidos para obtener un resultado y luego enviar el resultado por correo electrónico, se deberán crear al menos tres funciones: dos commandos y una consulta:

```
customers = read_customers()
customers_customers_with_debt = select_customers_with_debt(customers)
send_mail(customers_customers_with_debt)
```

**Muchos problemas que requieren recorridos (for, while) se pueden resolver empleando list comprehensions**

````python
def contains_errors(data):
    if isinstance(data, list):
        for item in data:
            if isinstance(item, str):
                if item == "error":
                    return True
    return False
````

````python
def contains_errors(data):
    if not isinstance(data, list):
        return False

    return "error" in data
````

````python
contains_errors([]) == False
````

````python
contains_errors(["ok", "fine"]) == False
````

````python
contains_errors(["ok", "error", "fine"]) == True
````

**Predicados/condiciones complejas.**

Para mejorar la legibilidad en el siguiente ejemplo se puede aplicar [Consolidate Conditional Expression](https://refactoring.com/catalog/consolidateConditionalExpression.html)

````python
def calcular_bono(empleado):
    bono = 10
    if empleado["tipo_empleado"] == "de_planta" and (empleado["años_servicio"] > 10 or \
                                       (empleado["estado_civil"] != "casado" and empleado["nro_hijos_menores_de_edad"] >= 3 )):
        bono = 50
    return bono
````

````python
def calcular_bono(empleado):
    bono = 10
    if satisface_criterios_bonificacion(empleado):
        bono = 50
    return bono

def satisface_criterios_bonificacion(empleado):
    return empleado["tipo_empleado"] == "de_planta" and (empleado["años_servicio"] > 10 or \
                                       (empleado["estado_civil"] != "casado" and empleado["nro_hijos_menores_de_edad"] >= 3 ))
````

````python
empleado = {
    "tipo_empleado": "consultor"
    , "años_servicio" : 15
    , "estado_civil" : "viudo"
    , "nro_hijos_menores_de_edad": 3
}


calcular_bono(empleado)
````
