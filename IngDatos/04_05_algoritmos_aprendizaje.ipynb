# Ranking práctico y pedagógico

> “Complejidad” aquí combina:
>
> * dificultad matemática,
> * costo computacional,
> * dificultad de entrenamiento,
> * capacidad de modelado,
> * cantidad de datos requerida,
> * sofisticación arquitectónica.

# 1. Regresión Lineal

* Predice valores continuos.
* Modelo más simple y explicable.
* Supone relación lineal entre variables.
* Muy rápido de entrenar.

### Usos

* Predicción de precios
* Tendencias
* Estadística clásica

### Ventajas

* Fácil de interpretar
* Muy eficiente

### Limitaciones

* No captura relaciones complejas

# 2. Regresión Logística

* Clasificación binaria.
* Base de muchos sistemas de clasificación.

### Usos

* Spam detection
* Diagnóstico médico
* Riesgo crediticio

### Complejidad

Solo un poco más compleja que regresión lineal.

# 3. k-Nearest Neighbors (k-NN)

* No aprende una función explícita.
* Clasifica según los vecinos más cercanos.

### Idea

> “Dime quiénes son tus vecinos y te diré quién eres.”

### Ventajas

* Muy intuitivo
* Sin entrenamiento pesado

### Problemas

* Muy lento en datasets grandes
* Escala mal dimensionalmente

# 4. Árboles de Decisión

* Divide los datos usando reglas.

Ejemplo:

```text
¿Edad < 18?
 ├── Sí → Niño
 └── No
      ├── ¿IMC > 30?
      └── Adulto
```

### Ventajas

* Muy interpretables
* Manejan relaciones no lineales

### Problemas

* Sobreajuste (overfitting)

# 5. Random Forest

* Conjunto de muchos árboles de decisión.

### Idea

Muchos árboles “votan” por la respuesta final.

### Potencia

Gran salto respecto a árboles simples.

### Muy usado en:

* Finanzas
* Medicina
* Sistemas tabulares

### Ventajas

* Robusto
* Excelente baseline

# 6. Gradient Boosting / XGBoost

Modelos famosos:

* XGBoost
* LightGBM
* CatBoost

### Idea

Cada árbol corrige los errores del anterior.

### Potencia

Durante años fue el rey de datos tabulares.

### Ventajas

* Rendimiento enorme
* Ganador frecuente en Kaggle

### Problemas

* Mucha ingeniería de hiperparámetros

# 7. Support Vector Machines (SVM)

* Busca el hiperplano óptimo de separación.

### Conceptualmente

Muy elegante matemáticamente.

### Potencia

Excelente en datasets medianos y complejos.

### Problemas

* Costoso computacionalmente
* Difícil de escalar

# 8. Redes Neuronales Feedforward (MLP)

* Primera entrada al deep learning.

### Estructura

```text
Entrada → Capas ocultas → Salida
```

### Potencia

Puede aproximar funciones extremadamente complejas.

### Problemas

* Requiere tuning
* Riesgo de overfitting
* Necesita GPUs en casos grandes

# 9. Redes Convolucionales (CNN)

* Especializadas en visión computacional.

### Idea

Detectan patrones jerárquicos:

* bordes,
* texturas,
* formas,
* objetos.

### Revolucionaron:

* reconocimiento facial,
* medicina por imágenes,
* vehículos autónomos.

### Complejidad

Muy alta:

* convoluciones,
* pooling,
* backpropagation profundo.

# 10. Transformers

Base de:

* OpenAI GPT
* Google Gemini
* Meta Llama

### Revolución

Transformaron:

* NLP
* visión
* audio
* robótica
* multimodal AI

### Potencia

Actualmente el paradigma dominante de IA avanzada.

### Características

* Atención (attention)
* Paralelización masiva
* Escalamiento extremo

### Problemas

* Coste computacional gigantesco
* Requieren enormes datasets
* Entrenamiento distribuido

# Resumen visual de complejidad

```text
Regresión Lineal
   ↓
Regresión Logística
   ↓
k-NN
   ↓
Árboles de Decisión
   ↓
Random Forest
   ↓
XGBoost
   ↓
SVM
   ↓
MLP
   ↓
CNN
   ↓
Transformers
```

# Observación importante

“Más complejo” no siempre significa “mejor”.

En problemas tabulares reales:

* XGBoost frecuentemente supera deep learning.

En lenguaje e imágenes:

* Transformers dominan ampliamente.

En datasets pequeños:

* modelos simples suelen generalizar mejor.


# Regla práctica moderna

| Tipo de problema         | Mejor enfoque típico      |
| ------------------------ | ------------------------- |
| Datos tabulares          | XGBoost / Random Forest   |
| NLP                      | Transformers              |
| Imágenes                 | CNN o Vision Transformers |
| Series temporales        | Transformers o boosting   |
| Problemas interpretables | Árboles o regresión       |
| Sistemas gigantes de IA  | Transformers              |
