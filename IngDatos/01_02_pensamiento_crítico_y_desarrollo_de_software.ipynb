# Pensamiento crítico

La capacidad de pensar de forma clara y racional y de entender las conexiones lógicas entre las ideas

Conceptos clave de la definición:
- **Claridad:** Evitar la ambigüedad y la confusión.

- **Racionalidad:** Basarse en la razón y la evidencia en lugar de las emociones puras.

- **Conexiones lógicas:** Identificar cómo un argumento lleva a otro o cómo se relacionan diferentes conceptos.


## Los Pilares del Pensamiento Crítico

---

### 1. Claridad

* Saber exactamente qué problema estás intentando resolver.
* La vaguedad es el enemigo de un buen análisis.

### 2. Lógica

* Estructurar tu razonamiento.
* Asegurarse de que tus conclusiones se deriven de tu evidencia.

### 3. Creatividad

* No conformarse con la primera hipótesis.
* Generar alternativas y explorar múltiples ángulos.



# Teoría del Pensamiento Dual de Kahneman

La teoría del pensamiento dual, popularizada por el Nobel **Daniel Kahneman** en su libro *Pensar rápido, pensar despacio*, propone que nuestro cerebro utiliza dos "sistemas" distintos para procesar la información y tomar decisiones.

Aunque no son regiones físicas del cerebro, funcionan como dos personajes con personalidades muy diferentes que viven dentro de nuestra cabeza.

---

## 1. El Sistema 1: El Pensamiento Rápido

Es el sistema **automático, intuitivo y emocional**. Funciona de manera involuntaria y no requiere esfuerzo consciente. Es el que nos permite sobrevivir en el día a día sin agotarnos mentalmente.

* **Características:** Veloz, asociativo y siempre encendido.
* **Ejemplos:** * Completar la frase "pan y...".
  * Poner cara de asco al ver una imagen desagradable.
  * Detectar que una persona está enojada por su tono de voz.
  * Moverse cuando algo se acerca a la cara.
  * La primera impresión cuando conocemos a alguien por primera vez.


## 2. El Sistema 2: El Pensamiento Despacio

Es el sistema **lógico, planificador y lento**. Requiere esfuerzo, concentración y consume mucha energía. Solo se activa cuando el Sistema 1 no encuentra una respuesta inmediata o cuando detectamos algo que rompe nuestro modelo del mundo.

* **Características:** Perezoso (prefiere no trabajar si no es necesario), analítico y racional.
* **Ejemplos:**
  * Mantener un paso más rápido de lo normal de forma consciente.
  * Buscar a una mujer con pelo blanco en una multitud.
  * Llenar un formulario de impuestos.
  * Revisar un contrato.
  * Diseñar software (gracias a la AI, no tanto programar).

---

## Conceptos clave para entender la teoría

### La "Ley del Mínimo Esfuerzo"

Kahneman explica que el cerebro es ahorrador de energía. Como el **Sistema 2 consume mucha glucosa y cansa**, tendemos a dejar que el **Sistema 1 tome el mando** la mayor parte del tiempo. El problema es que el Sistema 1 es propenso a cometer errores sistemáticos.

### Sesgos y Heurísticas

Debido a que el Sistema 1 es impulsivo, utiliza atajos mentales (llamados **heurísticas**) para llegar a conclusiones rápidas. Esto genera **sesgos cognitivos**: errores de juicio que cometemos una y otra vez.

> **Un ejemplo clásico:** Si un bate y una pelota cuestan `$1.10` en total, y el bate cuesta `$1.00` más que la pelota, ¿cuánto cuesta la pelota?
> El Sistema 1 grita inmediatamente: **"¡10 centavos!"**. Pero si activas el Sistema 2, te das cuenta de que la respuesta correcta es **5 centavos**.

---

## ¿Por qué es relevante esta teoría?

Entender estos dos sistemas nos ayuda a ser más conscientes de nuestras limitaciones:

1. **Toma de decisiones:** Nos enseña a "encender" el Sistema 2 en momentos críticos (finanzas, salud, relaciones, ejercicio profesional, aprendizaje) para evitar impulsos erróneos.
2. **Publicidad y Marketing:** Las marcas suelen hablarle a tu Sistema 1 (emociones, colores, deseos inmediatos) para que compres sin pensar.
3. **Empatía:** Entendemos que muchas de nuestras reacciones viscerales no son "nosotros", sino un sistema automático que podemos aprender a regular.


# Sesgos cognitivos

Los sesgos cognitivos son **"atajos mentales"** o errores sistemáticos de pensamiento que ocurren cuando el Sistema 1 (el intuitivo) toma el control y procesa la información de forma incompleta, distorsionada o dándole un peso exesivo a experiencias previas o influencias sociales.

"Nunca me he resfriado y tampoco me he aplicado la vacuna, por los tanto las vacunas no son necesarias"

Aunque existen cientos de ellos, Daniel Kahneman y otros psicólogos han identificado algunos que afectan nuestras decisiones todos los días:

---

### ¿Cómo evitarlos?

Kahneman advierte que **es casi imposible eliminarlos**, porque el Sistema 1 es demasiado rápido. Sin embargo, podemos mitigarlos al:

1. **Hacer pausas:** Obligar al Sistema 2 a revisar la decisión.
2. **Buscar opiniones contrarias:** Combatir el sesgo de confirmación activamente.
3. **Usar datos y estadísticas:** No confiar en lo que "sentimos" que es frecuente o probable.

## 1. Sesgo de Confirmación

Tendencia a buscar, interpretar y recordar información que confirme nuestras creencias previas, ignorando la evidencia contraria.

- **Sesgo:** Confirmación
- **Descripción:** Buscar solo datos que respalden nuestra hipótesis.
- **Perjuicio en Software:** El desarrollador solo hace pruebas unitarias que sabe que pasarán ("happy path"), ignorando casos de borde o errores potenciales.
- **Cómo evitar:** Fomentar el **Testing Destructivo** y buscar activamente pruebas que intenten "romper" el código en lugar de validarlo.


## 2. Sesgo de Anclaje

La tendencia a confiar demasiado en la primera pieza de información ofrecida (el "ancla") al tomar decisiones.


- **Sesgo:** Anclaje
- **Descripción:** Depender de la primera cifra o dato escuchado.
- **Perjuicio en Software:**  En una estimación de esfuerzo, si el líder dice "esto parece de 2 días", el equipo ajustará sus estimaciones reales hacia ese número, aunque sea irreal.
- **Cómo evitar:** Usar técnicas de estimación ciega como el **Planning Poker**, donde nadie conoce la cifra de los demás hasta el final.

## 3. Heurística de Disponibilidad

Dar más importancia a la información que recordamos con mayor facilidad o que es más reciente.

- **Sesgo:** Disponibilidad
- **Descripción:** Sobrestimar la frecuencia de eventos recientes o impactantes.
- **Perjuicio en Software:**  Elegir una arquitectura o librería solo porque se usó en el proyecto anterior o se vio en un blog ayer, sin evaluar si es la mejor opción técnica.
- **Cómo evitar:** Utilizar **cuadros comparativos** y matrices de decisión basadas en requerimientos objetivos, no en tendencias actuales.


## 4. Exceso de Confianza

Una sobreestimación de nuestras propias habilidades, conocimientos o la precisión de nuestras predicciones.

- **Sesgo:** Exceso de confianza
- **Descripción:** Creer que somos más hábiles de lo que realmente somos.
- **Perjuicio en Software:**  Ignorar la necesidad de documentación o revisiones de código porque "el código es perfecto" o subestimar drásticamente los plazos de entrega.
- **Cómo evitar:** Utilizar **cuadros comparativos** y matrices de decisión basadas en requerimientos objetivos, no en tendencias actuales.


## 5. Efecto Dunning-Kruger

Individuos con escasa habilidad en un área sufren de un sentimiento de superioridad ilusorio, considerando que son más inteligentes que los demás.

- **Sesgo:** Dunning-Kruger
- **Descripción:** Menos capacidad implica menos conciencia de la propia incompetencia.
- **Perjuicio en Software:** Un desarrollador junior intenta implementar un sistema complejo desde cero sin entender los riesgos de seguridad o escalabilidad, rechazando consejos.
- **Cómo evitar:** Utilizar **cuadros comparativos** y matrices de decisión basadas en requerimientos objetivos, no en tendencias actuales.

## 6. Efecto de Encuadre (Framing)

Tomar decisiones basadas en cómo se presenta la información, más que en la información misma.

- **Sesgo:** Encuadre
- **Descripción:** Reaccionar diferente según si algo se presenta como ganancia o pérdida.
- **Perjuicio en Software:** Un equipo acepta un cambio riesgoso si se dice "tenemos 90% de éxito", pero lo rechaza si se dice "hay 10% de probabilidad de tumbar producción".
- **Cómo evitar:** Replantear los problemas desde **múltiples perspectivas** (positiva y negativa) antes de tomar una decisión final.

## 7. Sesgo de Supervivencia

Concentrarse en las personas o cosas que superaron un proceso y pasar por alto a las que no, debido a su falta de visibilidad.

- **Sesgo:** Supervivencia
- **Descripción:** Analizar solo los éxitos y no los fallos.
- **Perjuicio en Software:** Copiar la arquitectura de Netflix o Google pensando que eso garantiza el éxito, sin ver los miles de startups que fallaron usando esa misma tecnología.
- **Cómo evitar:** Realizar **Post-mortems** de proyectos fallidos y estudiar por qué ciertas tecnologías o procesos no funcionaron en otros contextos.


## 8. Sesgo de Costo Hundido

Continuar con un proyecto o inversión debido a los recursos ya invertidos, incluso si las nuevas pruebas sugieren que fallará.

- **Sesgo:** Costo Hundido
- **Descripción:** Negarse a abandonar un camino erróneo por lo ya invertido.
- **Perjuicio en Software:** Seguir manteniendo un módulo de software obsoleto y propenso a errores solo porque "costó 6 meses programarlo", en lugar de refactorizar.
- **Cómo evitar:** Evaluar decisiones basadas en el **valor futuro**, no en el gasto pasado. Estar dispuesto a "matar" funciones que no aportan valor.

## 9. Efecto de Halo

Permitir que un rasgo positivo de una persona o herramienta influya en nuestra opinión sobre sus otras cualidades.


- **Sesgo:** Efecto de Halo
- **Descripción:** Generalizar a partir de una sola característica positiva.
- **Perjuicio en Software:** Creer que porque un desarrollador es excelente en Frontend, automáticamente será un buen Arquitecto de Sistemas o Líder de Equipo.
- **Cómo evitar:** Evaluar las competencias mediante **pruebas específicas** para cada rol o tecnología, separando el carisma de la habilidad técnica.

## 10. Sesgo de Pensamiento de Grupo (Groupthink)

Priorizar el consenso y la armonía del grupo sobre el pensamiento crítico y la evaluación de alternativas.

- **Sesgo:** Pensamiento de Grupo
- **Descripción:** Suprimir opiniones disidentes para evitar conflictos.
- **Perjuicio en Software:** El equipo no señala fallos críticos en un diseño porque el arquitecto principal está muy entusiasmado y nadie quiere "romper el ambiente".
- **Cómo evitar:** "Abogado del Diablo" en las reuniones de diseño técnico cuya misión sea cuestionar todas las decisiones.

## 11. Sesgo de Retrospectiva (Hindsight Bias)

La tendencia a ver eventos pasados como predecibles en el momento en que ocurrieron ("Yo ya lo sabía").

- **Sesgo:** Retrospectiva
- **Descripción:** Creer falsamente que un error era obvio después de que ocurrió.
- **Perjuicio en Software:** Culpar duramente a un desarrollador por un bug en producción diciendo "era evidente", generando miedo a innovar o tomar riesgos necesarios.
- **Cómo evitar:** Documentar el **proceso de decisión en tiempo real** (ADRs - Architecture Decision Records) para entender el contexto original de la decisión.

# Falacias lógicas

Mientras que los sesgos cognitivos son errores psicológicos del Sistema 1 (cómo sentimos o procesamos), **las falacias lógicas son errores en el razonamiento que hacen que un argumento parezca válido o persuasivo cuando en realidad no lo es**. Son patrones de pensamiento defectuosos que comprometen la solidez del razonamiento, aunque superficialmente puedan sonar convincentes.

Las falacias son peligrosas porque explotan cómo funciona naturalmente nuestra mente. Nos hacen aceptar conclusiones sin evidencia adecuada, nos distraen del verdadero debate, y permiten que argumentos débiles parezcan fuertes. Reconocerlas te ayuda a pensar más claramente y a no ser manipulado.

El Sistema 2 es el encargado de detectar estas trampas, pero como es perezoso, a menudo dejamos pasar argumentos falsos si la conclusión nos "suena" bien.

## Diferencia clave: Sesgo vs. Falacia

| Concepto | Origen | Definición |
| --- | --- | --- |
| **Sesgo Cognitivo** | Sistema 1 (Intuición) | Un "instinto" o atajo mental que nos inclina a un error. |
| **Falacia Lógica** | Sistema 2 (Razonamiento) | Un fallo en la construcción lógica de un discurso. |

> **Nota:** Un argumento puede ser una falacia incluso si la conclusión es verdadera. Por ejemplo: "El cielo es azul porque a los pitufos les gusta ese color". El cielo *es* azul, pero la lógica para llegar a esa conclusión es una falacia.

Entender las falacias lógicas es fundamental en la ingeniería de software, donde las decisiones deben basarse en datos, arquitectura y lógica sólida en lugar de argumentos retóricos.

## 1. El Hombre de Paja

* **Descripción:** Distorsionar o simplificar el argumento de otra persona para que sea más fácil de atacar.
* **Perjuicio en el software:** Durante una revisión de diseño, alguien propone: "Deberíamos agregar pruebas unitarias a este módulo". La respuesta falaz sería: "No podemos detener todo el desarrollo por meses solo para escribir pruebas perfectas". Se caricaturiza la propuesta como algo extremo para descartarla.
* **Cómo evitar:** Practica el **"Steelstepping"** (argumento de acero): resume la postura de tu colega de la manera más fuerte y fiel posible antes de intentar refutarla.


## 2. Ad Hominem

* **Descripción:** Atacar a la persona que hace el argumento en lugar de atacar el argumento en sí.
* **Perjuicio en el software:** Descartar una sugerencia de refactorización simplemente porque viene de un desarrollador junior o de alguien que cometió un error en el pasado. "No le hagas caso a Juan sobre la base de datos, el mes pasado rompió el servidor de staging".
* **Cómo evitar:** Separa el código (o la idea) del autor. Enfócate exclusivamente en los datos técnicos y los méritos de la propuesta.


## 3. Pendiente Resbaladiza

* **Descripción:** Afirmar que un pequeño primer paso conducirá inevitablemente a una cadena de eventos catastróficos.
* **Perjuicio en el software:** "Si permitimos que este equipo use una librería de Python diferente, pronto cada desarrollador usará un lenguaje distinto y el sistema será imposible de mantener en seis meses".
* **Cómo evitar:** Evalúa cada paso por sus propios méritos. Pide evidencia de por qué el "paso A" llevará necesariamente al "paso Z" sin posibilidad de control intermedio.

## 4. Falso Dilema (Falsa dicotomía)

* **Descripción:** Presentar solo dos opciones como si fueran las únicas posibles, ignorando alternativas intermedias.
* **Perjuicio en el software:** "O reescribimos toda la plataforma desde cero o seguiremos teniendo errores para siempre". Ignora opciones como el refactorizado gradual o la mejora de la cobertura de pruebas.
* **Cómo evitar:** Siempre pregunta: "¿Qué otras opciones tenemos?". Fomenta el pensamiento lateral y busca soluciones híbridas.

## 5. Razonamiento Circular

* **Descripción:** La conclusión ya está incluida en la premisa.
* **Perjuicio en el software:** "Este framework es el mejor porque es el estándar de la industria, y es el estándar de la industria porque es el mejor". No ofrece una razón técnica real (rendimiento, seguridad, curva de aprendizaje).
* **Cómo evitar:** Busca métricas externas y comparativas objetivas que no dependan de la afirmación original.

## 6. Post Hoc Ergo Propter Hoc

* **Descripción:** Asumir que porque un evento ocurrió después de otro, el primero fue la causa del segundo.
* **Perjuicio en el software:** "Ayer instalamos el nuevo plugin del IDE y hoy el servidor de producción se cayó. El plugin causó la caída". Esto puede llevar a cacerías de brujas innecesarias en lugar de un análisis de causa raíz real.
* **Cómo evitar:** Realiza un **Análisis de Causa Raíz (RCA)** basado en logs y telemetría, no solo en la secuencia temporal.

## 7. Apelación a la Autoridad

* **Descripción:** Sostener que algo es cierto solo porque una figura de autoridad lo dice.
* **Perjuicio en el software:** "Debemos usar Microservicios porque el CTO de Netflix dijo en una conferencia que es la única forma de escalar". Esto ignora si el contexto de tu empresa es similar al de Netflix.
* **Cómo evitar:** Valora el conocimiento experto, pero exige que la recomendación se adapte a las necesidades técnicas específicas de tu proyecto actual.

## 8. Apelación a la Emoción

* **Descripción:** Intentar ganar un argumento manipulando los sentimientos del interlocutor (miedo, culpa, orgullo).
* **Perjuicio en el software:** "Si no lanzamos esta funcionalidad el viernes, decepcionaremos a todo el equipo que ha trabajado duro y los inversores pensarán que no somos capaces". Se ignora si el código está listo o es seguro.
* **Cómo evitar:** Toma decisiones basadas en la definición de "Listo" (Definition of Done) y en la gestión de riesgos técnicos, no en la presión emocional.

## 9. Apelación a la Popularidad (Ad Populum)

* **Descripción:** Validar un punto de vista porque "todo el mundo lo hace".
* **Perjuicio en el software:** Adoptar una tecnología "hype" (como ciertas criptos o frameworks de moda) solo porque es tendencia en redes sociales, sin evaluar si resuelve un problema real en tu stack.
* **Cómo evitar:** Evalúa la tecnología mediante una Prueba de Concepto (PoC) y revisa su madurez y soporte a largo plazo, más allá de su popularidad.

## 10. Ningún escocés verdadero

* **Descripción:** Cambiar la definición de un grupo para excluir un contraejemplo que refuta tu argumento.
* **Perjuicio en el software:** * Persona A: "Ningún desarrollador senior dejaría pasar este error de seguridad".
* Persona B: "Pero el arquitecto principal lo aprobó".
* Persona A: "Entonces no es un *verdadero* desarrollador senior".
* Esto impide aprender de los errores y crea expectativas irreales.


* **Cómo evitar:** Define criterios claros y objetivos para los procesos y roles, y acepta que los errores son parte del sistema, independientemente del rango.
