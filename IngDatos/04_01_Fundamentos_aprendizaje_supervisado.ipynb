# Aprendizaje Automático (Machine Learning)

## Inteligencia Artificial, Aprendizaje Automático y Aprendizaje automático.

**Inteligencia Artificial (IA)**: El concepto general que abarca cualquier tecnología que permite a las máquinas realizar tareas que normalmente requerirían inteligencia humana, como el razonamiento, la percepción y la adaptación.

**Aprendizaje Automático (ML)**: Una aplicación específica de la IA que dota a las computadoras de la capacidad de aprender y mejorar a partir de la experiencia, utilizando algoritmos y modelos estadísticos para analizar datos e identificar patrones, sin ser programadas explícitamente para cada tarea.

**Aprendizaje Supervisado**: Es el tipo de aprendizaje automático más común, en el cual un algoritmo se entrena con un conjunto de datos etiquetados (pares de entrada y salida). El modelo aprende la relación entre las entradas y las salidas para hacer predicciones precisas sobre datos nuevos y no vistos

## ¿Qué es aprender para una máquina?

En las máquinas, "aprender" tiene un sentido matemático y computacional.

Un sistema "aprende" cuando mejora su rendimiento en una tarea específica mediante la experiencia (datos).

Formalmente (según Tom Mitchell, 1997):

*Un programa de computadora aprende de la experiencia **E** con respecto a una tarea **T** y una medida de desempeño **P**, si su desempeño en **T**, medido por **P**, mejora con la experiencia *E*.*

## Aprendizaje Automático vs Sistemas Basados en Reglas

### ⚙️ 1. Sistemas basados en reglas (Rule-Based Systems)
**🧩 Definición:**

Son sistemas que siguen un conjunto explícito de reglas definidas por humanos para tomar decisiones.
Cada regla se expresa en forma de “SI ... ENTONCES ...” (if-then).

**💡 Características:**

- Las reglas son escritas manualmente por expertos.
- No aprenden de los datos.
- l conocimiento está explícitamente programado.
- Son fáciles de entender y depurar.
- Escalan mal cuando hay muchos casos o excepciones (difícil mantener miles de reglas).

**✅ Ventajas:**
- Transparencia (sabemos por qué toma una decisión).
- Buena precisión en dominios muy bien definidos.
- No necesita grandes cantidades de datos.

**❌ Desventajas:**
- No aprende por sí mismo.
- Muy rígido ante nuevos casos no previstos.
- Mantenimiento costoso si las reglas cambian.

### 🤖 2. Aprendizaje Supervisado
**🧠 Definición:**

Es un enfoque del aprendizaje automático (Machine Learning) donde el sistema aprende automáticamente las reglas a partir de datos etiquetados.

El modelo analiza muchos ejemplos y descubre por sí solo los patrones o "reglas"  que mejor explican los datos.

**💡 Características:**
- Requiere datos etiquetados (ejemplos con la respuesta correcta).
- Usa algoritmos (como regresión, árboles de decisión, redes neuronales).
- Puede generalizar a nuevos casos.
- Las "reglas" son aprendidas matemáticamente (no escritas).

**✅ Ventajas:**
- Aprende automáticamente de los datos.
- Se adapta a nuevos contextos.
- Puede manejar problemas muy complejos (visión, voz, texto).

**❌ Desventajas:**
- Necesita muchos datos de calidad.
- Menos interpretable ("caja negra").
- Puede sobreajustarse si no se controla bien.

### ⚖️ Comparación directa:

| Aspecto                     | Sistemas Basados en Reglas           | Aprendizaje Supervisado           |
| --------------------------- | ------------------------------------ | --------------------------------- |
| **Fuente del conocimiento** | Reglas escritas por expertos humanos | Patrones aprendidos de los datos  |
| **Forma de decisión**       | Lógica “SI-ENTONCES” explícita       | Modelo matemático o estadístico   |
| **Aprendizaje**             | No aprende                           | Aprende con datos etiquetados     |
| **Adaptabilidad**           | Rígido                               | Flexible y generalizable          |
| **Transparencia**           | Alta (explicable)                    | Baja (a veces “caja negra”)       |
| **Datos necesarios**        | Pocos o ninguno                      | Muchos y bien etiquetados         |
| **Ejemplo típico**          | Sistema experto médico (años 80-90)  | Clasificador de imágenes o textos |
